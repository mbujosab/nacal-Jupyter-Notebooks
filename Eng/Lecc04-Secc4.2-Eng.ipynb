## Pivots and pre-echelon matrices



Read the beginning of [Section 4.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.4.2). The pivot of a nonzero column is its first nonzero component. The book calls pivot position of that column the index of the row in which its pivot is. A matrix is a pre-echelon matrix if the components to the right of each pivot are zero. The following matrix is the example in the book.

## Elimination from left to right



Read [Section 4.2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.2.1). To reduce a matrix to pre-echelon form is to make the components to the right of each pivot zero by means of elementary operations. The columns are processed from left to right. The following matrix is the example in the book.



```python
A = Matrix([ [0,1,1], [2,-2,2], [2,-8,-4] ])
A
```

The pivot of the first column is the 2 in the second row. To its right there are a $-2$ and a 2. To make them zero, the first column is added to the second and the first column is subtracted from the third. The operations are applied to a copy of $\Mat{A}$, called `M`.



```python
M = Matrix(A)
M & T( (1,1,2) ) & T( (-1,1,3) )
```

Now the pivot of the second column is the 1 in the first row. To its right there is a 1. The second column is subtracted from the third.



```python
M & T( (-1,2,3) )
```

The matrix is a pre-echelon matrix. Its third column is zero. If you run one of these cells twice, the operation is applied twice. In that case run again the cell that defines `M` and repeat the steps.

`Elim` does the elimination from left to right. With a 1 as second argument it shows the steps.



```python
Elim(A, 1)
```

The steps are the same. `Elim` does not change the matrix $\Mat{A}$: it returns a new matrix.

### Fractions



In the following matrix the pivots are not ones.



```python
C = Matrix([ [2,1,3], [3,5,-1], [4,-2,7] ])
Elim(C, 1)
```

The pivot of the first column is a 2 and to its right there is a 1. To make it zero with a single operation you would have to subtract half the first column from the second column. NAcAL avoids the fraction: first it multiplies the second column by 2 and then it subtracts the first column from it. Multiplying a column by 2 is an operation of Type II. The book explains it after Theorem 4.2.1.

The proof of Theorem 4.2.1 makes the components zero with operations of Type I. NAcAL also has that variant. The last part of this Notebook explains the variants.



```python
C.elim(1, rep=True)
```

The two matrices obtained are pre-echelon forms of $\Mat{C}$ and they are not equal. A matrix has many pre-echelon forms. Compare the pivots of the two: are they in the same rows?

Reduce $\Mat{C}$ to pre-echelon form on paper in the way you prefer. Then check your steps with the copy `N`. In NAcAL a fraction is written with `fracc`: `fracc(1,2)` is one half.



```python
N = Matrix(C)
```

## Gauss-Jordan elimination



Read [Section 4.2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.2.3). A matrix in reduced echelon form is an echelon matrix in which each pivot is a 1 and the other components in the row of each pivot are zero.

The next cell stores in `L` an echelon form of $\Mat{B}$.



```python
L = ElimG(B)
L
```

The proof of Corollary 4.2.3 reduces it in this order. You start with the pivot of the first column. With each pivot the components to its left are made zero. At the end each column is divided by its pivot. The first pivot has nothing to its left. The second pivot is the 1 in the second row and to its left there is a 2.



```python
L & T( (-2,2,1) )
```

The third pivot is the 5 in the third row. Make the two components to its left zero. Then divide the third column by 5. You will need `fracc`.



`ElimGJ` does Gauss-Jordan elimination.



```python
ElimGJ(B, 1)
```

The final matrix is the same as yours, but the steps are not the same. The book warns that the order of the operations is not fixed. NAcAL avoids fractions and leaves the reordering of the columns for the end. The last part of this Notebook deals with that.

The next cell applies Gauss-Jordan elimination to the matrix $\Mat{A}$ defined at the beginning.



```python
ElimGJ(A, 1)
```

### The variants of elimination



`Elim`, `ElimG` and `ElimGJ` use the same procedure, called `elim`, which has variants. Each variant has a number. To combine several of them you add their numbers.


| Number|What it does|
|---|---|
| 0|Reduces to pre-echelon form from left to right and avoids fractions, like <code>Elim</code>|
| +1|Uses only operations of Type I, even if fractions appear|
| +2|Also makes the components to the left of each pivot zero|
| +10|Divides each column by its pivot|
| +20|Reorders the columns so that the matrix is an echelon matrix|
| +100|Processes the columns from right to left|

With `rep=True` the steps are shown. Variant 2 makes the components on both sides of each pivot zero.



```python
B.elim(2, rep=True)
```

The sum $2+10+20=32$ is Gauss-Jordan elimination.



```python
B.elim(32, rep=True)
```

Write the variant that uses only operations of Type I and also reorders the columns. Apply it to the matrix `A2` of the part about the problems.



Variant 100 goes through the columns from right to left. In each column it uses as pivot the last nonzero component and makes zero the components that this pivot has to its left.



```python
B.elim(100, rep=True)
```

The result is not a pre-echelon matrix with the definition in the book, because here the pivot is a different one. Count its nonzero columns.

Reduce $\Mat{B}$ to pre-echelon form with variants 0, 1, 2 and 20 and count the nonzero columns in each case. There are always three. Lecture 5 proves that all the pre-echelon forms of a matrix have the same number of pivots. It calls that number the rank of the matrix. With $\Mat{B}$ variants 0 and 1 give the same matrix. Why?



### Chaining stages



`elim` returns a matrix, and another variant can be applied to that matrix. In this way you decide the order of the stages. The next cell follows the order of the proof of Corollary 4.2.3. It reduces to pre-echelon form. It reorders the columns. It makes zero the components to the left of the pivots. It divides each column by its pivot.



```python
B.elim(0).elim(20).elim(2).elim(10, rep=True)
```

Compare with the steps of `ElimGJ(B, 1)`. The order of the steps is different and the final matrix is the same.

Change the order of the stages and run it again. Try for example to divide by the pivots before reordering the columns.

## What you should know at the end



-   The pivot of a nonzero column is its first nonzero component.
-   A matrix is a pre-echelon matrix if the components to the right of each pivot are zero. It is an echelon matrix if, in addition, the zero columns are on the right and each pivot is lower than that of the previous column. It is in reduced echelon form if, in addition, each pivot is a 1 and the other components in its row are zero.
-   Every matrix can be reduced to pre-echelon form, to echelon form and to reduced echelon form with a sequence of elementary column operations (Theorem 4.2.1 and Corollaries 4.2.2 and 4.2.3).
-   To reduce to pre-echelon form by hand you process the columns from left to right. With the pivot of each column you make the components to its right zero. A zero column is left as it is.
-   A matrix has many pre-echelon forms. NAcAL avoids fractions with operations of Type II, and that is why its steps may differ from the ones you take on paper.
-   `Elim`, `ElimG` and `ElimGJ` are for checking what you have first done on paper.
