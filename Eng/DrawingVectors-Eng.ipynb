## The `draw` function



NAcAL does not draw. The next cell defines a function, `draw`, that shows vectors of $\R[2]$ as arrows. It uses the [Matplotlib](https://matplotlib.org/) library. You do not need to understand its code. Run the cell and go on to the next section.



```python
import matplotlib.pyplot as plt

def draw(*vectors, chained=False, names=None):
    """Draws vectors of R^2 as arrows.

    draw(a, b, c)   All the arrows start at the origin.

    draw(a, b, c, chained=True)
        Each arrow starts at the tip of the previous one. The sum
        of all the vectors is also drawn, in black.

    draw(a, b, names=['a', 'b'])
        Changes the text of the legend. Without this argument the
        legend shows the components of each vector."""
    for v in vectors:
        if not isinstance(v, Vector) or v.n != 2:
            raise ValueError('draw only accepts vectors of R^2. '
                             'Options are written with their name: chained=True')
    colors = ['tab:blue', 'tab:red', 'tab:green', 'tab:orange', 'tab:purple', 'tab:brown']
    fig, ax = plt.subplots()

    def arrow(start, end, color, name):
        ax.annotate('', xy=end, xytext=start,
                    arrowprops=dict(arrowstyle='->', color=color, lw=2))
        ax.plot([], [], color=color, lw=2, label=name)

    def text(x, y):
        return '(' + str(x) + ', ' + str(y) + ',)'

    points = [(0, 0)]
    start = (0, 0)
    total = (0, 0)
    for k, v in enumerate(vectors):
        end = (start[0] + float(1|v), start[1] + float(2|v))
        name = names[k] if names else text(1|v, 2|v)
        arrow(start, end, colors[k % len(colors)], name)
        points.append(end)
        total = (total[0] + (1|v), total[1] + (2|v))
        if chained:
            start = end
    if chained:
        arrow((0, 0), start, 'black', 'sum: ' + text(total[0], total[1]))

    xs = [p[0] for p in points]
    ys = [p[1] for p in points]
    ax.set_xlim(min(xs) - 1, max(xs) + 1)
    ax.set_ylim(min(ys) - 1, max(ys) + 1)
    ax.set_aspect('equal')
    ax.grid(True)
    ax.axhline(0, color='gray', lw=1)
    ax.axvline(0, color='gray', lw=1)
    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1))
    plt.show()
```

## Vectors from the origin



The horizontal axis is the first component and the vertical axis is the second. Each vector is an arrow that starts at the origin and ends at the point given by its components.



```python
a = Vector([1, 2])
b = Vector([3, -1])
draw(a, b)
```

The legend shows the components of each vector. The `names` argument changes that text.



```python
draw(a, b, names=['a', 'b'])
```

Now draw $2\Vect{a}$, $\Vect{a}$ and $-\Vect{a}$, in that order, in the same figure. With that order the arrow of $\Vect{a}$ lies on top of the arrow of $2\Vect{a}$ and can be seen. Look at what the three arrows have in common.



Draw $\Vect{a}$, $\Vect{b}$ and $\Vect{a}+\Vect{b}$ in the same figure.



## Addition: one arrow after another



With `chained=True` each arrow starts at the tip of the previous one. The black arrow is the sum. It starts at the origin and ends where the last arrow ends.



```python
draw(a, b, chained=True)
```

Change the order of the two vectors and repeat the drawing. Look at where the black arrow ends in both cases. Which property of vector addition are you seeing?



Subtracting a vector is adding its opposite. Draw $\Vect{a}-\Vect{b}$ this way: chain $\Vect{a}$ and $-\Vect{b}$.



## Linear combinations



A linear combination of $\Vect{a}$ and $\Vect{b}$ is a sum of multiples of those vectors. This drawing shows $2\Vect{a}+3\Vect{b}$.



```python
draw(2*a, 3*b, chained=True)
```

Change the two coefficients until the sum is the vector $(6,\,5,)$. Use the legend to see where the black arrow ends.



The coefficients can be negative or fractions. To write the fraction $\frac{1}{2}$ use `fracc(1,2)`. Draw $\frac{1}{2}\Vect{a}-2\Vect{b}$.



## The column picture of a system of equations



This way of looking at a system is explained in class. Consider the system

\begin{equation*}
\begin{cases}
x+3y=6 \\
2x-y=5
\end{cases}
\end{equation*}

The coefficients of $x$ form the vector $(1,\,2,)$. The coefficients of $y$ form the vector $(3,\,-1,)$. The numbers on the right-hand side form the vector $(6,\,5,)$. They are the vectors $\Vect{a}$ and $\Vect{b}$ of this Notebook. The system asks which linear combination of $\Vect{a}$ and $\Vect{b}$ is equal to $(6,\,5,)$:

\begin{equation*}
x\,\Vect{a}+y\,\Vect{b}=\begin{pmatrix}6\\5\end{pmatrix}.
\end{equation*}

You found those coefficients in the previous section. Check with NAcAL, without drawing, that the linear combination gives the vector $(6,\,5,)$. Check it also as the product of a matrix and a vector: the matrix is `Matrix([a, b])`.

## What you should know at the end



-   A vector of $\R[2]$ is drawn as an arrow from the origin.
-   The multiples of a vector lie on the same line through the origin.
-   To add vectors, each arrow is placed after the previous one. The order does not change the result.
-   Solving a system of two equations with two unknowns is finding the linear combination of its two columns that gives the right-hand side vector.
