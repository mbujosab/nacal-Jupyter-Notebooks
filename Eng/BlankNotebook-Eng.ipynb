## Reminder of commands



A vector.



```python
a = Vector([1, 2, 3])
a
```

A matrix, written row by row.



```python
A = Matrix([[1, 2, 3], [2, 4, 7], [1, 1, 1]])
A
```

The selector on the right picks a column of the matrix.



```python
A|2
```

The selector on the left picks a row.



```python
1|A
```

With both selectors you pick a component.



```python
1|A|2
```

A fraction.



```python
Vector([fracc(1, 3), 2])
```

Symbolic variables.



```python
x, y = sympy.symbols('x y')
Vector([x, y, 1])
```

Elimination to a pre-echelon form. With the second argument the steps are shown.



```python
Elim(A, 1)
```

Gaussian elimination, to an echelon form.



```python
ElimG(A, 1)
```

Gauss-Jordan elimination, to the reduced echelon form.



```python
ElimGJ(A, 1)
```

The inverse matrix.



```python
A.inversa()
```

The rank.



```python
A.rango()
```
