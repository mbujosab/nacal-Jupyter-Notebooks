## Product of a vector and a matrix



Read the definition at the beginning of [Section 2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.3). The product $\Vect{b}\Mat{A}$ is the linear combination of the rows of $\Mat{A}$ whose coefficients are the components of $\Vect{b}$. The matrix and the vector are those of the previous Notebook.



```python
A = Matrix([ Vector([1,1,1]), Vector([1,0,1]), Vector([-2,2,1]) ])
b = Vector([1, 2, -3])
A
```

The first cell writes the linear combination of the rows. The second writes the product.



```python
(b|1)*(1|A) + (b|2)*(2|A) + (b|3)*(3|A)
```

```python
b * A
```

The two cells give the same vector. Now compute $\Mat{A}\Vect{b}$ and compare it with $\Vect{b}\Mat{A}$.



The two products are different. The first one combines the columns of $\Mat{A}$ and the second one combines its rows.

## A vector times a matrix is the transpose of the matrix times the vector



The rows of $\Mat{A}$ are the columns of its transpose. That is why $\Vect{b}\Mat{A}=(\MatT{A})\Vect{b}$. The book proves it in [Section 2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.3), after the definition. In NAcAL the transpose of `A` is `~A`.



```python
b * A == (~A) * b
```

The book then states the equality for the other product: $\Mat{A}\Vect{b}=\Vect{b}(\MatT{A})$. It comes from applying the previous equality to the matrix $\MatT{A}$, because the transpose of $\MatT{A}$ is $\Mat{A}$. Check it.



## Each component is a dot product



The $j$-th component of the vector $\Vect{b}\Mat{A}$ is the dot product of $\Vect{b}$ with the $j$-th column of $\Mat{A}$.



```python
(b*A)|3
```

```python
b * (A|3)
```

The two cells give the same number. Check the same for the second component.



## Properties of the product of a vector and a matrix



Read [Section 2.3.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.3.1). Its exercise states six properties. Three are written here. Write the other three yourself. The cells also use this matrix and this vector.



```python
B = Matrix([ [1,2,0], [0,1,3], [2,2,1] ])
c = Vector([2, 1, -1])
```

First: $\Vect{b}\Mat{I}=\Vect{b}$.



```python
b * I(3) == b
```

Second: the $i$-th row of the identity times a matrix is the $i$-th row of the matrix.



```python
(2|I(3)) * A == 2|A
```

Third: $(\Vect{b}+\Vect{c})\Mat{A}=\Vect{b}\Mat{A}+\Vect{c}\Mat{A}$.



Fourth: $(\lambda\Vect{b})\Mat{A}=\lambda(\Vect{b}\Mat{A})$.



Fifth: $\Vect{b}(\lambda\Mat{A})=(\lambda\Vect{b})\Mat{A}$.



```python
b * (3*A) == (3*b) * A
```

Sixth: $\Vect{b}(\Mat{A}+\Mat{B})=\Vect{b}\Mat{A}+\Vect{b}\Mat{B}$.



The book deduces all six from the properties of the product of a matrix and a vector. It uses the equality $\Vect{b}\Mat{A}=(\MatT{A})\Vect{b}$.



## What you should know at the end



-   The product $\Vect{b}\Mat{A}$ is a linear combination of the rows of $\Mat{A}$. The coefficients are the components of $\Vect{b}$.
-   $\Mat{A}\Vect{b}$ and $\Vect{b}\Mat{A}$ are different vectors in general. If the matrix is not square, only one of the two products can exist with the same vector.
-   $\Vect{b}\Mat{A}=(\MatT{A})\Vect{b}$ and $\Mat{A}\Vect{b}=\Vect{b}(\MatT{A})$.
-   The $j$-th component of $\Vect{b}\Mat{A}$ is the dot product of $\Vect{b}$ with the $j$-th column of $\Mat{A}$.
-   The product of a vector and a matrix is linear in the vector and linear in the matrix.
