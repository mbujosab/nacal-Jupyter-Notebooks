## Product of a matrix and a vector



Read [Section 2.2.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.2.2) up to the first exercise. The product $\Mat{A}\Vect{b}$ is the linear combination of the columns of $\Mat{A}$ whose coefficients are the components of $\Vect{b}$. The following matrix and vector are those of the example in the book.



```python
A = Matrix([ Vector([1,1,1]), Vector([1,0,1]), Vector([-2,2,1]) ])
b = Vector([1, 2, -3])
A
```

The first cell writes the linear combination of the columns. The second writes the product.



```python
(A|1)*(b|1) + (A|2)*(b|2) + (A|3)*(b|3)
```

```python
A * b
```

The two cells give the same vector. The first one is the definition of the product.

Write the linear combination $2\Vect{x}+2\Vect{y}+2\Vect{z}$ of the previous section as the product of a matrix and a vector. The matrix whose columns are those three vectors is `Matrix([x, y, z])`.



Now write the linear combination $2\,\VectC{A}{1}-\VectC{A}{3}$ as the product of a matrix and a vector. Compute it first with the columns and compare the two results.



The matrix $\Mat{A}$ has three columns. The vector must have three components, one for each column. Run the next cell and read the last line of the error. It is the message of the previous Notebook. Here it says that the matrix and the vector cannot be multiplied.



```python
A * Vector([1, 2])
```

The following matrix has two rows and three columns. The vector has three components, one for each column.



```python
C = Matrix([ [1,0,2], [3,1,1] ])
C * Vector([1, 2, 3])
```

The product has two components. It is a linear combination of the columns, which are vectors of $\R[2]$. The product $\Mat{A}\Vect{b}$ has as many components as $\Mat{A}$ has rows.

### Each component is a dot product



Read the rest of [Section 2.2.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.2.2). The $i$-th component of the vector $\Mat{A}\Vect{b}$ is the dot product of the $i$-th row of $\Mat{A}$ with $\Vect{b}$.



```python
2|(A*b)
```

```python
(2|A) * b
```

The two cells give the same number. The first one selects the second component of the product. The second one multiplies the second row of $\Mat{A}$ by $\Vect{b}$. Check the same for the first component.



Python performs multiplication before selection. That is why `2|A*b` is the same as `2|(A*b)`. The number is the same as that of `(2|A)*b`: it is the equality you have just checked. The output does not tell in which order Python has operated. The next two cells do show it.



```python
A*b|2
```

```python
A*(b|2)
```

The first cell computes $\Mat{A}\Vect{b}$ and selects its second component. The second cell multiplies the matrix by the number $b_2$.

If the matrix has only one row, the product is a vector of $\R[1]$. It is not a number.



```python
F = Matrix([ [1, 2, 3] ])
F * b
```

```python
(1|F) * b
```

The first cell gives a vector with one component. The second gives a number.



## Properties of the product of a matrix and a vector



Read [Section 2.2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.2.3). The following cells also use this matrix and this vector.



```python
B = Matrix([ [1,2,0], [0,1,3], [2,2,1] ])
c = Vector([2, 1, -1])
B
```

### Linear in the vector



Distributive with respect to addition: $\Mat{A}(\Vect{b}+\Vect{c})=\Mat{A}\Vect{b}+\Mat{A}\Vect{c}$.



```python
A * (b + c) == A*b + A*c
```

Associative with respect to scalar multiplication: $\Mat{A}(\lambda\Vect{b})=\lambda(\Mat{A}\Vect{b})$. Write it yourself.



With the scalar on the right, this property is $\Mat{A}(\Vect{b}\lambda)=(\Mat{A}\Vect{b})\lambda$. Only the parentheses move.



```python
A * (b*3) == (A*b) * 3
```

### A matrix times the product of another matrix and a vector



The corollary of [Section 2.2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.2.3) says that $\Mat{A}(\Mat{B}\Vect{c})$ is the product of a new matrix and $\Vect{c}$. The columns of that matrix are the products of $\Mat{A}$ and each column of $\Mat{B}$.



```python
A * (B * c)
```

```python
Matrix([ A*(B|1), A*(B|2), A*(B|3) ]) * c
```

The two cells give the same vector. Show the matrix of the second cell without multiplying it by $\Vect{c}$. Compare each column with `A*(B|1)`, `A*(B|2)` and `A*(B|3)`. In Lecture 3 this matrix will be the product $\Mat{A}\Mat{B}$.



### Linear in the matrix



Distributive with respect to addition: $(\Mat{A}+\Mat{B})\Vect{c}=\Mat{A}\Vect{c}+\Mat{B}\Vect{c}$.



```python
(A + B) * c == A*c + B*c
```

Associative with respect to scalar multiplication: $(\lambda\Mat{A})\Vect{b}=\lambda(\Mat{A}\Vect{b})$. Write it yourself.



The next cell is the proposition of the exercise in the book: $\Mat{A}(\lambda\Vect{b})=(\lambda\Mat{A})\Vect{b}$. With the previous equality it gives $(\lambda\Mat{A})\Vect{b}=\Mat{A}(\lambda\Vect{b})=\lambda(\Mat{A}\Vect{b})$.



```python
A * (3*b) == (3*A) * b
```

### The scalar can go anywhere



```python
3 * A * b
```

```python
A * 3 * b
```

```python
A * b * 3
```

The three cells give the same vector. No parentheses are needed. What does not change is that the vector is to the right of the matrix.



## What you should know at the end



-   A linear combination of vectors is a sum of multiples of those vectors.
-   The product $\Mat{A}\Vect{b}$ is a linear combination of the columns of $\Mat{A}$. The coefficients are the components of $\Vect{b}$.
-   The number of components of $\Vect{b}$ is the number of columns of $\Mat{A}$. The product has as many components as $\Mat{A}$ has rows.
-   $\Mat{I}\Vect{a}=\Vect{a}$. A matrix times the $j$-th column of the identity is the $j$-th column of the matrix.
-   The $i$-th component of $\Mat{A}\Vect{b}$ is the dot product of the $i$-th row of $\Mat{A}$ with $\Vect{b}$.
-   The product of a matrix and a vector is linear in the vector and linear in the matrix.
-   $\Mat{A}(\Mat{B}\Vect{c})$ is the product with $\Vect{c}$ of the matrix whose columns are the products of $\Mat{A}$ and each column of $\Mat{B}$.
-   The scalar can go anywhere in the product. The vector always goes to the right of the matrix.
-   Python performs multiplication before selection: `A*b|2` is the second component of `A*b`.
