## Product of matrices



Read the definition at the beginning of [Section 3.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.1) and the example that follows it. The $j$-th column of $\Mat{AB}$ is the product of $\Mat{A}$ and the $j$-th column of $\Mat{B}$. The following matrices are those of the example in the book.



```python
x = Vector([1, 1, -2])
y = Vector([1, -2, 1])
z = Vector([-2, 1, 1])
A = Matrix([x, y, z])
B = Matrix([z, x])
A
```

```python
B
```

The first cell writes the product of $\Mat{A}$ and the second column of $\Mat{B}$. The second cell writes the second column of the product $\Mat{AB}$.



```python
A * (B|2)
```

```python
(A*B)|2
```

The two cells give the same vector. Check the same for the first column.



Now write the whole product and compare its two columns with the previous vectors.



```python
A * B
```

Each column of $\Mat{AB}$ is a linear combination of the columns of $\Mat{A}$. The coefficients are the components of the corresponding column of $\Mat{B}$.

Several numbers are repeated in the product of the book. That is why a mistake can give the right result. The following two matrices give a product whose components are all different.



```python
R = Matrix([ [1,2,0], [3,-1,4], [0,1,3] ])
S = Matrix([ [2,1], [0,-1], [1,3] ])
R * S
```

Write the second column of $\Mat{RS}$ as a linear combination of the three columns of $\Mat{R}$.



### When the product exists



The product is only defined if the first matrix has as many columns as the second one has rows. $\Mat{A}$ has three columns and $\Mat{B}$ has three rows. That is why $\Mat{AB}$ exists. Decide whether $\Mat{BA}$ exists before running the next cell.



```python
B * A
```

NAcAL gives an error, because $\Mat{B}$ has two columns and $\Mat{A}$ has three rows.



## Properties of the product of matrices



Read the two exercises of [Section 3.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.1) that state Propositions 3.1.1 to 3.1.7. The cells of this part also use these matrices and this vector. $\Mat{C}$ has the same order as $\Mat{A}$, and $\Mat{D}$ has the same order as $\Mat{B}$.



```python
C = Matrix([ [1,0,2], [0,1,-1], [3,1,0] ])
D = Matrix([ [0,1], [2,-1], [1,3] ])
F = Matrix([ [2,1], [-1,3] ])
v = Vector([1, 2])
```

### The product is associative



Proposition 3.1.1: $\Mat{A}(\Mat{B}\Vect{v})=(\Mat{AB})\Vect{v}$.



```python
A * (B*v)
```

```python
(A*B) * v
```

The two cells give the same vector. Proposition 3.1.2: $\Mat{A}(\Mat{BF})=(\Mat{AB})\Mat{F}$. Check it.



That is why the book writes $\Mat{ABF}$ without parentheses. Python computes `A*B*F` from left to right, and it gives the same matrix.



```python
A * B * F
```

### The product is linear



Proposition 3.1.3: $(\Mat{A}+\Mat{C})\Mat{B}=\Mat{AB}+\Mat{CB}$.



```python
(A + C) * B == A*B + C*B
```

Proposition 3.1.4: $\Mat{A}(\Mat{B}+\Mat{D})=\Mat{AB}+\Mat{AD}$. Write it yourself.



Proposition 3.1.5: $\Mat{A}(\lambda\Mat{B})=\lambda(\Mat{AB})=(\lambda\Mat{A})\Mat{B}$. The cell checks the first equality with $\lambda=5$. Write the second one yourself.



```python
A * (5*B) == 5 * (A*B)
```

The book sums up these three propositions like this: the product of matrices is a linear operator, and it is also linear on the left.



### The identity matrix



Propositions 3.1.6 and 3.1.7: $\Mat{IB}=\Mat{B}$ and $\Mat{BI}=\Mat{B}$. $\Mat{B}$ has three rows and two columns. That is why the identity in the first equality has order 3.



```python
I(3) * B == B
```

What is the order of the identity in the second equality? Check it.



## The product of matrices is not commutative



Read the paragraph «Muy muy importante» of [Section 3.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.1). In general $\Mat{AB}$ is different from $\Mat{BA}$. The following two matrices are those of the book.



```python
G = Matrix([ [1,1], [0,1] ])
H = Matrix([ [1,0], [0,3] ])
G * H
```

```python
H * G
```

The two matrices differ in one component. With matrices that are not square, one of the two products may exist and the other may not. That is what happened with $\Mat{BA}$ at the beginning of this Notebook.

The book gives two more properties of numbers that matrices do not have. Among numbers, if $\lambda^2=\lambda$ then $\lambda$ is 0 or 1. The following matrix is equal to its square and it is neither the zero matrix nor the identity.



```python
P = Matrix([ [1,0], [0,0] ])
P * P == P
```

Among numbers, if $\lambda^2=0$ then $\lambda$ is 0. Compute the square of the following matrix.



```python
N = Matrix([ [1,1], [-1,-1] ])
N
```

The square is the zero matrix, and $\Mat{N}$ is not.



## Two other ways of computing the product



Read [Section 3.1.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.3.1.1). Its two propositions follow from the definition of the product by columns.



### Component by component



Proposition 3.1.8: the component in row $i$ and column $j$ of $\Mat{AB}$ is the dot product of the $i$-th row of $\Mat{A}$ with the $j$-th column of $\Mat{B}$. The cells use the matrices $\Mat{R}$ and $\Mat{S}$.



```python
2|(R*S)|1
```

```python
(2|R) * (S|1)
```

The two cells give the same number. Check the same for the component in row 3 and column 2.



This is the usual way of computing the product by hand: rows times columns.



### Row by row



Proposition 3.1.9: the $i$-th row of $\Mat{AB}$ is the product of the $i$-th row of $\Mat{A}$ and the matrix $\Mat{B}$.



```python
2|(R*S)
```

```python
(2|R) * S
```

The two cells give the same vector. Check the same for the third row.



Therefore each row of $\Mat{AB}$ is a linear combination of the rows of $\Mat{B}$. The coefficients are the components of the corresponding row of $\Mat{A}$. Write the second row of $\Mat{RS}$ as a linear combination of the three rows of $\Mat{S}$.



The book asks you to verify the two propositions with the product of its example. It is the product $\Mat{AB}$ at the beginning of this Notebook. Compute it by hand in the three ways: column by column, component by component and row by row. Compare each result with the output of the cell `A * B`.



## What you should know at the end



-   The $j$-th column of $\Mat{AB}$ is $\Mat{A}$ times the $j$-th column of $\Mat{B}$. The columns of $\Mat{AB}$ are linear combinations of the columns of $\Mat{A}$.
-   The product only exists if $\Mat{A}$ has as many columns as $\Mat{B}$ has rows.
-   The product is associative. It is a linear operator and it is also linear on the left.
-   $\Mat{IA}=\Mat{A}$ and $\Mat{AI}=\Mat{A}$, with the identity of the right order on each side.
-   In general $\Mat{AB}$ is different from $\Mat{BA}$.
-   The component in row $i$ and column $j$ of $\Mat{AB}$ is the dot product of the $i$-th row of $\Mat{A}$ with the $j$-th column of $\Mat{B}$.
-   The $i$-th row of $\Mat{AB}$ is the $i$-th row of $\Mat{A}$ times $\Mat{B}$. The rows of $\Mat{AB}$ are linear combinations of the rows of $\Mat{B}$.
