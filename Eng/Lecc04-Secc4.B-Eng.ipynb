## The same elementary matrix from a row operation or from a column operation



Read the beginning of [Section 4.B](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph2). The same elementary matrix of Type I can be obtained by operating on the columns of the identity or by operating on its rows. The two operations are not the same: their indices are interchanged.



```python
I(3) & T( (5,2,3) )
```

```python
T( (5,3,2) ) & I(3)
```

Both cells give the same matrix. The first one adds 5 times the second column to the third. The second one adds 5 times the third row to the second.

## The mirror of an elementary operation



The book calls mirror («espejo») of an operation the operation that gives the same result when it acts on the other side of the identity matrix. In NAcAL it is obtained with `.espejo()`. The mirror of an operation of Type I has the two indices interchanged.



```python
T( (5,2,3) ).espejo()
```

The operations of Type II and the interchanges are their own mirror, because their matrices are symmetric.



```python
T( (4,2) ).espejo()
```

```python
T( {2,3} ).espejo()
```

Check with the identity matrix of order 3 that the previous interchange gives the same matrix as a row operation and as a column operation.



## The mirror of a sequence



The mirror of a sequence of operations is the sequence of the mirrors, in the same order. The following sequence is the example in the book.



```python
Tr = T( (3,2,1) ) & T( (-2,3,2) ) & T( (7,1,3) ) & T( {2,3} ) & T( (20,1) )
Tr
```

```python
Tr.espejo()
```

Only the indices of the three operations of Type I have changed. The book proves that the mirror applied to the rows of the identity gives the same matrix as the sequence applied to its columns.



```python
I(3) & Tr
```

```python
Tr.espejo() & I(3)
```

### The order in which they act



The two sequences have the operations in the same order, but they do not act in the same order. On the columns the operation on the left acts first. On the rows the one closest to the matrix acts first, which is the last one of the sequence.

Apply the five steps of the mirror sequence to the rows of the matrix `M`, one in each cell. Start with the one that acts first. The first cell is already written. At the end you should get the matrix of the two previous cells.



```python
M = I(3)
T( (20,1) ) & M
```

### The mirror is not the transpose



Do not confuse the mirror of a sequence with its transpose, which is from the Notebook for Section 4.A. The mirror changes the indices and keeps the order. The transpose keeps the indices and reverses the order.



```python
~Tr
```

The transposed sequence applied to the rows of the identity gives the transpose of the matrix `I(3) & Tr`. Check it.



### With any matrix



Operating on the rows of a matrix is multiplying it on the left by the matrix obtained by operating on the rows of the identity. For the mirror sequence that matrix is ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$. Therefore, applying the mirror sequence to the rows of $\Mat{A}$ is multiplying $\Mat{A}$ on the left by ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$.



```python
A = Matrix([ [1,2,0], [3,-1,4], [2,5,1] ])
Tr.espejo() & Matrix(A)
```

```python
( I(3) & Tr ) * A
```

So the same matrix ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ multiplies on the right when the sequence is applied to the columns, and on the left when its mirror is applied to the rows. Check the first statement.



## Symmetric matrices and the mirror



Read the paragraph «Matrices simétricas y transformación espejo» of [Section 4.B](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph2). The matrix ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ is symmetric if and only if the mirror of the sequence and its transpose give the same matrix when they are applied to the rows of the identity. The following sequence is the example in the book.



```python
p = T( [ (9,2), (18,1), (-1,2,1), (fracc(1,9),1), (-1,1,2), (fracc(1,2),2) ], rpr='h' )
(p.espejo() & I(2)) == (~p & I(2))
```

```python
p.espejo() & I(2)
```

The two sequences give the same matrix, but they are not the same sequence. Compare the three.



```python
p
```

```python
p.espejo()
```

```python
~p
```

## The mirror of the inverse of an operation



Read the end of [Section 4.B](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph2). The inverse of an elementary operation is the operation that undoes it. In NAcAL it is written with `**-1`. Lecture 5 deals with inverse matrices.



```python
T( (3,1,2) )**-1
```

```python
T( (2,3) )**-1
```

The inverse of a sequence undoes the steps in the opposite order. Lecture 5 proves it.



```python
Tr**-1
```

Apply the sequence `Tr` to a copy of $\Mat{A}$ and then its inverse. You should get the matrix $\Mat{A}$ again.



The book deduces that, if a sequence is applied to the columns of the identity and the mirror of its inverse to its rows, you get the identity again.



```python
(Tr**-1).espejo() & I(3) & Tr
```

Lecture 17 applies those two sequences to any square matrix: the sequence to its columns and the mirror of the inverse to its rows. The matrix obtained is said to be similar to the first one. The next cell does it with a short sequence.



```python
S = T( [ (2,1,2), (-1,3,1) ] )
(S**-1).espejo() & Matrix(A) & S
```

Write the sequence `(S**-1).espejo()` and say what each of its steps does with the rows.



## Python details



-   `.espejo()` and `**-1` are applied to an operation or to a sequence and give another sequence. They do not change the original sequence.
-   The parentheses in `(Tr**-1).espejo()` are needed. Without them Python gives an error.
-   Python computes `**` before `&`. That is why `Tr**-1` does not need parentheses in `Matrix(A) & Tr & Tr**-1`.
-   In the cell `(Tr**-1).espejo() & I(3) & Tr` there is no copy, because `I(3)` creates a new identity matrix each time.



## What you should know at the end



-   The mirror of an operation is the operation that gives the same matrix when it acts on the other side of the identity.
-   The mirror of an operation of Type I has the indices interchanged. Those of Type II and the interchanges are their own mirror.
-   The mirror of a sequence is the sequence of the mirrors, in the same order. Applied to the rows of the identity it gives the same matrix as the sequence applied to the columns.
-   The mirror changes the indices and keeps the order. The transpose keeps the indices and reverses the order.
-   The matrix ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ is symmetric if and only if the mirror of the sequence and its transpose give the same matrix when they are applied to the rows of the identity.
-   If a sequence is applied to the columns of the identity and the mirror of its inverse to its rows, you get the identity again. Lecture 17 uses those two sequences to obtain similar matrices.
