## The dot product



Read the definition of the dot product at the beginning of [Section 2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.1). The dot product of two vectors of $\R[n]$ is a number: the sum of the products of their components. In NAcAL it is written with the symbol `*`.



```python
a = Vector([1, 2, 3])
b = Vector([1, -1, 2])
a * b
```

Compute the same number with the components of the two vectors. Multiply the first components, the second ones and the third ones, and add the three products. The first product is `(a|1)*(b|1)`.



The two vectors must have the same number of components. Run the next cell and read the error message.



```python
a * Vector([1, 2])
```

The message is in the last line: `ValueError: Sistema y Vector incompatibles`. It is in Spanish. It means that the two vectors cannot be multiplied: one has three components and the other has two. The lines before it show where in the code of NAcAL the error was detected. You do not need to read them.



## Properties of the dot product



The first exercise of [Section 2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.1) asks you to prove four properties. Each of the following cells compares the two sides of an equality with the vectors $\Vect{a}$, $\Vect{b}$ and $\Vect{c}$.



```python
c = Vector([2, 1, -1])
```

Symmetry: $\Vect{a}\cdot\Vect{b}=\Vect{b}\cdot\Vect{a}$.



```python
a * b == b * a
```

Linear in the first argument. Associative with respect to scalar multiplication: $(\lambda\Vect{a})\cdot\Vect{b}=\lambda(\Vect{a}\cdot\Vect{b})$.



```python
(3*a) * b == 3 * (a*b)
```

Distributive with respect to addition: $(\Vect{a}+\Vect{b})\cdot\Vect{c}=\Vect{a}\cdot\Vect{c}+\Vect{b}\cdot\Vect{c}$. Write it yourself.



Nonnegative: $\Vect{x}\cdot\Vect{x}\geq0$. Compute the dot product of $\Vect{b}$ with itself. The second component of $\Vect{b}$ is negative. Compute also `(b|2)*(b|2)`: it is the term that this component contributes.



Definite: $\Vect{x}\cdot\Vect{x}=0$ if and only if $\Vect{x}$ is the zero vector. Compute the dot product of the zero vector of $\R[3]$ with itself. The zero vector is `V0(3)`. Then try a vector with only one component different from zero. Can the result be 0?



These cells check each property for these vectors. They do not prove it for all vectors. The proof is in the book.



## The dot product is also linear in the second argument



The book deduces it from the symmetry. Check the two equalities: $\Vect{a}\cdot(\lambda\Vect{b})=\lambda(\Vect{a}\cdot\Vect{b})$ and $\Vect{a}\cdot(\Vect{b}+\Vect{c})=\Vect{a}\cdot\Vect{b}+\Vect{a}\cdot\Vect{c}$. Use $\lambda=3$.



Since the dot product is linear in both arguments, the scalar can multiply either of the two vectors.



```python
a * (3*b) == (3*a) * b
```

## Dot product with a row or a column of the identity matrix



Read the special case that closes [Section 2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.1). The identity matrix of order 3 is `I(3)`.



```python
(1|I(3)) * b
```

```python
b * (I(3)|2)
```

The first cell gives the first component of $\Vect{b}$. The second gives the second component. Now obtain the third component of $\Vect{b}$ with a dot product.



## The dot product operates on two vectors



Read in [Section 2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.1) the paragraph before the properties of the dot product. It says that $(\Vect{a}\cdot\Vect{b})\Vect{c}$ and $\Vect{a}(\Vect{b}\cdot\Vect{c})$ are different vectors in general. Run the two cells and compare the results.



```python
a * b * c
```

```python
a * (b * c)
```

The two cells give different vectors. In the first one, Python computes `a*b` first, which is a number, and multiplies that number by $\Vect{c}$. In the second one it computes `b*c` first and multiplies $\Vect{a}$ by that number. Neither of them is a dot product of three vectors. That product does not exist. Always write the parentheses.



## What you should know at the end



-   The dot product of two vectors of $\R[n]$ is a number.
-   The dot product is symmetric, linear in both arguments, nonnegative and definite.
-   The dot product of a vector with the $i$-th row or column of the identity matrix is the $i$-th component of the vector.
-   The dot product takes two vectors. $(\Vect{a}\cdot\Vect{b})\Vect{c}$ and $\Vect{a}(\Vect{b}\cdot\Vect{c})$ are different vectors in general.
-   Checking an equality with an example is not proving it.
