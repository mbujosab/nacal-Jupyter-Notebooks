## Three expressions that are written without parentheses



Read the beginning of [Section 3.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.2). The three expressions in the box can be read in two ways: as the selection of a component or as a dot product. Both readings give the same number. The cells of this Notebook use these matrices and these vectors.



```python
A = Matrix([ [1,2,0], [3,-1,1] ])
B = Matrix([ [2,1], [0,-1], [1,3] ])
a = Vector([2, 1, -1])
b = Vector([1, -2, 2])
c = Vector([3, -1])
```

First: the $i$-th component of $\Mat{A}\Vect{b}$ is the dot product of the $i$-th row of $\Mat{A}$ with $\Vect{b}$. You checked it in the Notebook for Section 2.2.



```python
2|(A*b)
```

```python
(2|A) * b
```

Second: the $j$-th component of $\Vect{a}\Mat{B}$ is the dot product of $\Vect{a}$ with the $j$-th column of $\Mat{B}$. Write the two readings with $j=2$.



Third: the component in row $i$ and column $j$ of $\Mat{AB}$ is the dot product of the $i$-th row of $\Mat{A}$ with the $j$-th column of $\Mat{B}$. It is Proposition 3.1.8 of the previous Notebook. Write the two readings with $i=1$ and $j=2$.



On paper the parentheses are not needed. The dot is written when, after putting the parentheses, two vectors are next to each other. In Python they are needed: Python computes the product before the selection, and `2|A*b` is the same as `2|(A*b)`. For the other reading you have to write `(2|A)*b`.



## A vector times a matrix times a vector



Proposition 3.2.1: $(\Vect{a}\Mat{B})\cdot\Vect{c}=\Vect{a}\cdot(\Mat{B}\Vect{c})$.



```python
(a*B) * c
```

```python
a * (B*c)
```

The two cells give the same number. That is why the book writes $\Vect{a}\Mat{B}\Vect{c}$ without parentheses. In that expression no two vectors are next to each other, and that is why it has no dot.



```python
a * B * c
```

Write another vector of $\R[3]$ and another one of $\R[2]$ and check the proposition again.



## Transpose of a product



Read the last exercise of [Section 3.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.2). Proposition 3.2.2: $(\Mat{AB})^\T=(\MatT{B})(\MatT{A})$. In NAcAL the transpose of `A` is `~A`.



```python
~(A*B)
```

```python
(~B) * (~A)
```

The two cells give the same matrix. Transposing changes the order of the factors. Try the other order: $(\MatT{A})(\MatT{B})$.



With these matrices the product exists, but it is a matrix of order 3. It cannot be the transpose of $\Mat{AB}$, which has order 2.

The book warns that the expression $\Mat{A}\MatT{B}$ is ambiguous. Always write the parentheses: `~(A*B)` is the transpose of the product and `A*(~B)` is $\Mat{A}$ times the transpose of $\Mat{B}$.



### Two symmetric matrices



The book deduces from Proposition 3.2.2 that $(\MatT{A})\Mat{A}$ and $\Mat{A}(\MatT{A})$ are symmetric for any matrix $\Mat{A}$.



```python
(~A) * A
```

```python
((~A) * A).es_simetrica()
```

Compute $\Mat{A}(\MatT{A})$ and ask NAcAL whether it is symmetric.



Both are symmetric, but they do not have the same order. What is the order of each one if $\Mat{A}$ has order $m$ by $n$?

## What you should know at the end



-   The expressions $\elemL{\Mat{A}\Vect{b}}{i}$, $\elemR{\Vect{a}\Mat{B}}{j}$ and $\elemR{\elemL{\Mat{AB}}{i}}{j}$ are written without parentheses, because their two readings give the same number.
-   $(\Vect{a}\Mat{B})\cdot\Vect{c}=\Vect{a}\cdot(\Mat{B}\Vect{c})$. That is why we write $\Vect{a}\Mat{B}\Vect{c}$, without parentheses and without a dot.
-   $(\Mat{AB})^\T=(\MatT{B})(\MatT{A})$: transposing a product changes the order of the factors.
-   $(\MatT{A})\Mat{A}$ and $\Mat{A}(\MatT{A})$ are symmetric for any matrix $\Mat{A}$.
-   In Python you have to write the parentheses that are omitted on paper.
