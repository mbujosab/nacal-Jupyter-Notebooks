## Elementary operations of Type I



Read the beginning of Lecture 4 and [Section 4.1.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.1). An elementary operation of Type I adds to a vector a multiple of another vector. In NAcAL it is written with `T` and three numbers in parentheses. The following one adds 8 times the second vector to the third.



```python
T( (8,2,3) )
```

The first number is the one that multiplies. The last one is the index of the vector that changes.

The operator `&` applies the operation to the columns of the matrix on its left. If that matrix is the identity, the result is an elementary matrix.



```python
I(3) & T( (8,2,3) )
```

Look at where the 8 is: in row 2 and column 3. They are the two indices of the operation, in the same order.

Write the elementary matrix of order 4 obtained by adding $-7$ times the first column to the third. Say first in which row and in which column the $-7$ will be.



The cells of this Notebook use these three matrices.



```python
A = Matrix([ [1,2,0], [3,-1,4] ])
B = Matrix([ [2,1], [0,-1], [1,3] ])
C = Matrix([ [0,1,5], [7,0,4] ])
A
```

The next cell adds 2 times the first column to the third. The operation is applied to `Matrix(A)`, which is a copy of $\Mat{A}$.



```python
Matrix(A) & T( (2,1,3) )
```

Compare with $\Mat{A}$. Only the third column has changed.



### The operation changes the matrix



The operator `&` does not create a new matrix: it changes the matrix on its left. That is why the previous cell uses a copy. The next three cells show it with another copy, called `D`.



```python
D = Matrix(A)
D & T( (2,1,3) )
```

```python
D
```

```python
A
```

The matrix `D` has changed and the matrix `A` has not. In this Notebook the operations are almost always applied to a copy. In this way the three matrices defined at the beginning can be used in all the cells. If you change one of these matrices by mistake, run again the cell that defines them.



## Elementary operations of Type II



Read [Section 4.1.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.2). An elementary operation of Type II multiplies a vector by a nonzero number. In NAcAL it is written with two numbers: the number that multiplies and the index of the vector. The following one multiplies the first vector by 5.



```python
T( (5,1) )
```

```python
I(3) & T( (5,1) )
```

The elementary matrix is like the identity, except for the first component of the diagonal.

Multiply by $-10$ the second column of a copy of $\Mat{A}$.



### What is not an elementary operation



Multiplying a column by zero is not an elementary operation. Subtracting a column from itself is not one either. NAcAL does not accept them.



```python
T( (0,2) )
```

NAcAL gives an error. Try `T( (-1,2,2) )`, which would subtract the second vector from itself.



## Operating on the columns is multiplying by an elementary matrix



Read [Section 4.1.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.3). The book proves that ${\Mat{A}}_{\boldsymbol{\tau}}=\Mat{A}\big({\Mat{I}}_{\boldsymbol{\tau}}\big)$. Applying an elementary operation to the columns of $\Mat{A}$ gives the same as multiplying $\Mat{A}$ by the elementary matrix of that operation.



```python
Matrix(A) & T( (2,1,3) )
```

```python
A * ( I(3) & T( (2,1,3) ) )
```

Both cells give the same matrix. The identity matrix has order 3 because $\Mat{A}$ has three columns.

Check the equality with the operation of Type II that multiplies the second column by 5.



### A new rewriting rule



From that equality the book deduces that ${\big(\Mat{A}\Mat{B}\big)}_{\boldsymbol{\tau}}=\Mat{A}\big({\Mat{B}}_{\boldsymbol{\tau}}\big)$. The operation in the next cells subtracts 3 times the first column from the second.



```python
(A*B) & T( (-3,1,2) )
```

```python
A * ( Matrix(B) & T( (-3,1,2) ) )
```

Both cells give the same matrix. That is why the book writes ${\Mat{A}\Mat{B}}_{\boldsymbol{\tau}}$ without parentheses. Compute $\Mat{AB}$ and say which column has changed.



## Elementary matrices and their notation



Read [Section 4.1.4](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.4). The notation of an operation does not say what the order of its elementary matrix is. The same operation gives an elementary matrix of each order.



```python
I(2) & T( (-7,1,2) )
```

```python
I(4) & T( (-7,1,2) )
```

The identity matrix is an elementary matrix of both types. It is obtained by adding zero times a column to another. It is also obtained by multiplying a column by 1. NAcAL accepts both operations.



```python
I(3) & T( (0,1,2) )
```

```python
I(3) & T( (1,2) )
```

The transpose of an elementary matrix is another elementary matrix of the same type. In those of Type I the two indices are interchanged.



```python
~( I(3) & T( (8,2,3) ) )
```

```python
I(3) & T( (8,3,2) )
```

## Sequences of elementary operations



Read [Section 4.1.5](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.5). With `&` you can also chain operations to form a sequence. They are applied from left to right. The next cell is the example in the book. First it multiplies the second column of $\Mat{C}$ by 3. Then it adds 2 times the first column to the second.



```python
Matrix(C) & T( (3,2) ) & T( (2,1,2) )
```

A sequence can be stored in a variable. It can also be written as a list inside `T`.



```python
Tr = T( [ (3,2), (2,1,2) ] )
Tr
```

```python
Matrix(C) & Tr
```

The order matters. Apply the same two operations to a copy of $\Mat{C}$ in the opposite order. Which component is different?



### A sequence is a product of elementary matrices



The matrix ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ is the product of the elementary matrices of the sequence, in the same order.



```python
I(3) & Tr
```

```python
( I(3) & T( (3,2) ) ) * ( I(3) & T( (2,1,2) ) )
```

Both cells give the same matrix. It is not an elementary matrix: it differs from the identity in two components.

The book deduces that ${\Mat{A}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}=\Mat{A}\big({\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}\big)$. Multiply $\Mat{C}$ by the previous matrix and compare with the matrix you obtained when you applied the sequence to $\Mat{C}$.



### A sequence is a linear operator



An exercise in Section 4.1.5 asks you to prove that applying a sequence of elementary operations to the columns is a linear operator. The proof is a task for pen and paper. Here it is checked in one case. First with the product by a scalar.



```python
(5*C) & Tr
```

```python
5 * ( Matrix(C) & Tr )
```

Now check that operating on the sum $\Mat{A}+\Mat{C}$ gives the same as adding the two matrices obtained by operating on each one.



## Interchanges



Read [Section 4.1.6](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.6). The interchange of two vectors is written in NAcAL with the two indices in braces.



```python
T( {1,3} )
```

```python
Matrix(A) & T( {1,3} )
```

An interchange matrix is obtained by interchanging two columns of the identity matrix.



```python
I(3) & T( {1,3} )
```

The book proves that every interchange matrix is symmetric and that its square is the identity matrix. Check both things with the previous matrix. For the first one use `.es_simetrica()`.



### The interchange is a sequence of elementary operations



The book does not count the interchange as an elementary operation, because it is achieved with three operations of Type I and one of Type II. Do on paper the first part of the exercise in Section 4.1.6. The next cell applies to the identity matrix the sequence of the solution in the book.



```python
I(3) & T( [ (1,1,2), (-1,2,1), (1,1,2), (-1,1) ] )
```

Each step is seen better in a matrix whose columns are different. The next cell takes the first step with a copy of $\Mat{A}$.



```python
D = Matrix(A)
D & T( (1,1,2) )
```

Take the other three steps, one in each cell. Each cell changes `D`, so you do not have to copy it again. At the end the first two columns of $\Mat{A}$ are interchanged. If you run a cell twice, the operation is applied twice. In that case run again the cell that defines `D` and start over.



## Permutations



Read [Section 4.1.7](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.7). A sequence of interchanges reorders the columns. Applied to the identity matrix it gives a permutation matrix.



```python
I(4) & T( [ {1,3}, {4,2}, {3,2} ] )
```

Interchange matrices are symmetric. Permutation matrices are not symmetric in general. The following one is the example in the book.



```python
P = I(3) & T( [ {1,3}, {1,2} ] )
P
```

```python
P.es_simetrica()
```

Multiply $\Mat{A}$ by $\Mat{P}$. Say first on paper to which position each column of $\Mat{A}$ will go.



## Python details



-   `T` takes a single argument. That is why there are two pairs of parentheses in `T( (8,2,3) )`. Three numbers in parentheses are an operation of Type I. Two numbers are one of Type II. Two indices in braces are an interchange.
-   A list of operations inside `T` is a sequence. `T( [ (3,2), (2,1,2) ] )` is the same as `T( (3,2) ) & T( (2,1,2) )`.
-   The operator `&` changes the matrix on its left. `Matrix(A)` creates a copy of the matrix `A`.
-   `A*B` and `~A` create new matrices. You do not need to copy them before operating on them.



## What you should know at the end



-   There are two types of elementary operations. Those of Type I add to a vector a multiple of another vector. Those of Type II multiply a vector by a nonzero number.
-   An elementary matrix is obtained by applying a single elementary operation to the columns of the identity matrix. Its transpose is another elementary matrix of the same type.
-   ${\Mat{A}}_{\boldsymbol{\tau}}=\Mat{A}\big({\Mat{I}}_{\boldsymbol{\tau}}\big)$: operating on the columns of a matrix is multiplying it on the right by an elementary matrix. That is why ${\big(\Mat{A}\Mat{B}\big)}_{\boldsymbol{\tau}}=\Mat{A}\big({\Mat{B}}_{\boldsymbol{\tau}}\big)$.
-   Applying a sequence of elementary operations is multiplying by a product of elementary matrices. The order of the operations matters.
-   Applying a sequence of elementary operations to the columns is a linear operator.
-   The interchange is a sequence of operations of Type I and one of Type II. Interchange matrices are symmetric and their square is the identity. Permutation matrices are not symmetric in general.
-   In NAcAL the operator `&` changes the matrix it acts on.
