## Triangular matrices



Read the three definitions at the beginning of [Section 3.A](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph1). A matrix is upper triangular if all the components below the main diagonal are zero. It is lower triangular if all the components above it are zero.



```python
T = Matrix([ [2,1,3], [0,4,5], [0,0,6] ])
T
```

```python
T.es_triangularSup()
```

The transpose of an upper triangular matrix is lower triangular. Check it with `.es_triangularInf()`.



A triangular matrix does not have to be square. The following one has two rows and three columns.



```python
R = Matrix([ [1,2,3], [0,4,5] ])
R.es_triangularSup()
```

The book says that a zero matrix is upper triangular and also lower triangular. In NAcAL the zero matrix of order 2 by 3 is `M0(2,3)`. Check both things.



A triangular matrix is unit triangular if all the components of its diagonal are equal to one. In NAcAL `.diag()` gives the diagonal. The matrix $\Mat{T}$ is not unit triangular.



```python
T.diag()
```

## Product of two upper triangular matrices



Read the exercise of [Section 3.A](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph1), which states Propositions 3.A.1 to 3.A.3. To see what happens with any matrix, the following two have letters instead of numbers.



```python
a, b, c, d, e, f = sympy.symbols('a b c d e f')
p, q, r, s, t, u = sympy.symbols('p q r s t u')
U = Matrix([ [a,b,c], [0,d,e], [0,0,f] ])
V = Matrix([ [p,q,r], [0,s,t], [0,0,u] ])
U
```

```python
V
```

Proposition 3.A.1: the product of two upper triangular matrices is upper triangular.



```python
U * V
```

Look at the components below the diagonal: they are zero. Then look at the diagonal. What relation do you see between the diagonal of $\Mat{UV}$ and the diagonals of $\Mat{U}$ and $\Mat{V}$?



### When one of the two is unit triangular



Propositions 3.A.2 and 3.A.3 require square matrices of the same order. The following matrix is $\Mat{U}$ with ones on the diagonal.



```python
U1 = Matrix([ [1,b,c], [0,1,e], [0,0,1] ])
U1
```

Proposition 3.A.2: if the first matrix is unit upper triangular and the second one is upper triangular, the product is upper triangular and it keeps the diagonal of the second one.



```python
U1 * V
```

The diagonal of the product is that of $\Mat{V}$. Proposition 3.A.3: if the first one is upper triangular and the second one is unit upper triangular, the product keeps the diagonal of the first one. Check it with $\Mat{V}$ and $\Mat{U1}$.



Corollary 3.A.4 says that the product of two unit upper triangular matrices, square and of the same order, is also unit upper triangular. Check it with the square of $\Mat{U1}$, which is `U1*U1`.



## Lower triangular matrices



The book says that the same results hold for lower triangular matrices, and that transposing is enough to prove it. The transposes of $\Mat{U}$ and $\Mat{V}$ are lower triangular. Multiply them and look at where the zeros are.



Compare your result with the transpose of $\Mat{VU}$. Remember Proposition 3.2.2: transposing a product changes the order of the factors.



## What you should know at the end



-   A matrix is upper triangular if the components below the main diagonal are zero. It is lower triangular if its transpose is upper triangular.
-   A triangular matrix is unit triangular if its diagonal consists of ones.
-   The product of two upper triangular matrices is upper triangular. The same holds for lower triangular matrices.
-   If both are square and of the same order and one of them is unit triangular, the product keeps the diagonal of the other one.
-   The product of two unit upper triangular matrices, square and of the same order, is unit upper triangular.
