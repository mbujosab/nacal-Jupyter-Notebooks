## Operating on the rows



Read the beginning of [Section 4.A](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph1). To operate on the rows of a matrix the book writes the operation on the left: ${}_{\boldsymbol{\tau}}\Mat{A}$. In NAcAL it is also written on the left. The cells of this Notebook use these two matrices.



```python
A = Matrix([ [1,2,0], [3,-1,4], [2,5,1] ])
B = Matrix([ [2,1], [0,-1], [1,3] ])
A
```

The next cell adds 2 times the first row to the third. As in the Notebook for Section 4.1, the operation is applied to a copy. If you change one of the two matrices by mistake, run again the cell that defines them.



```python
T( (2,1,3) ) & Matrix(A)
```

The same operation, written on the right, adds 2 times the first column to the third.



```python
Matrix(A) & T( (2,1,3) )
```

The book arrives at row operations through the transpose. You operate on the columns of $\MatT{A}$ and transpose the result.



```python
~( (~A) & T( (2,1,3) ) )
```

It is the same matrix given by the first of the three cells.



### Operating on the rows is multiplying on the left



The book proves that ${}_{\boldsymbol{\tau}}\Mat{A}=\big({}_{\boldsymbol{\tau}}\Mat{I}\big)\Mat{A}$. Operating on the rows of $\Mat{A}$ gives the same as multiplying $\Mat{A}$ on the left by the matrix obtained by operating on the rows of the identity.



```python
T( (2,1,3) ) & I(3)
```

```python
( T( (2,1,3) ) & I(3) ) * A
```

Check the equality with the matrix $\Mat{B}$ and the operation that multiplies the second row by 5. What must the order of the identity matrix be?



### The elementary matrix for the rows is the transpose of the one for the columns



The book also proves that ${}_{\boldsymbol{\tau}}\Mat{I}={\big({\Mat{I}}_{\boldsymbol{\tau}}\big)}^\T$.



```python
T( (2,1,3) ) & I(3)
```

```python
I(3) & T( (2,1,3) )
```

Look at where the 2 is in each matrix. When you operate on the columns it is in row 1 and column 3, as in the Notebook for Section 4.1. When you operate on the rows it is in row 3 and column 1.

Apply an operation of Type II to the rows of the identity and then to its columns. Why do you get the same matrix?



## Sequences of row operations



In a sequence written on the left of the matrix, the operation closest to the matrix acts first. The next cell first multiplies the second row by 2 and then interchanges the first two rows.



```python
T( {1,2} ) & T( (2,2) ) & Matrix(A)
```

Write the same two operations in the other order and say, before running it, which row will end up multiplied by 2.



### The transpose of a sequence



The book calls transpose of a sequence the sequence with the same operations in the opposite order. In NAcAL it is written with `~`. The following sequence is the example in the book.



```python
Tr = T( [ (66,1,2), {3,2}, (4,1) ], rpr='h' )
Tr
```

```python
~Tr
```

The sequence applied to the columns of the identity and its transpose applied to the rows give two matrices. Each one is the transpose of the other.



```python
I(3) & Tr
```

```python
~Tr & I(3)
```

Now apply the sequence without transposing it to the rows of the identity. You get neither of the two previous matrices.



With any matrix the same happens as with a product of matrices. When you transpose, the sequence changes sides and is transposed: ${\big({\Mat{A}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}\big)}^\T={}_{(\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k)^\T}\big(\MatT{A}\big)$.



```python
~( Matrix(A) & Tr )
```

```python
~Tr & (~A)
```

### The notation is still associative



The book highlights in a box that $\big({}_{\boldsymbol{\tau}}\Mat{A}\big)\Mat{B}={}_{\boldsymbol{\tau}}\big(\Mat{A}\Mat{B}\big)$. Check it with the sequence `Tr` and the matrices $\Mat{A}$ and $\Mat{B}$.



## Symmetric matrices



Read the paragraph «Matrices simétricas» of [Section 4.A](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph1). The matrix ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ is symmetric if and only if it is equal to the matrix obtained by applying the transposed sequence to the rows of the identity. The following sequence is the example in the book.



```python
p = T( [ (9,2), (18,1), (-1,2,1), (fracc(1,9),1), (-1,1,2), (fracc(1,2),2) ], rpr='h' )
p
```

```python
(~p & I(2)) == (I(2) & p)
```

```python
I(2) & p
```

Make the same comparison with the sequence `Tr`. Then compute the matrix and say whether it is symmetric.



## Row echelon matrix



Read [Section 4.A.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.alph3.4.Alph1.1). A matrix is a row echelon matrix if its transpose is an echelon matrix. The pivot of a nonzero row is its first nonzero component. `ElimGF` reduces to row echelon form with row operations. The following matrix is the example in the book.



```python
G = Matrix([ [0,2,2], [1,-2,-8], [3,6,-6] ])
ElimGF(G, 1)
```

The row operations are written below the arrows. When there are several in one step, the bottom one acts first.

The book reduces to row echelon form in this way: it reduces the transpose to echelon form and transposes the result. Check it with `ElimG`, which is from the [Notebook for Section 4.2](Lecc04-Secc4.2-Eng.ipynb).



Now reduce the matrix $\Mat{A}$ to row echelon form on paper. Then apply your operations to the copy `M`. Remember that they are written on the left.



```python
M = Matrix(A)
```

```python
ElimGF(A, 1)
```

Your steps may differ from those of NAcAL. If you eliminate from top to bottom, the pivot of each row ends up in the same column as the one of NAcAL.



## Python details



-   With `rpr='h'` the sequence is shown horizontally. Without that argument it is shown vertically.
-   Python applies `~` before `&`. That is why `~Tr & I(3)` is the same as `(~Tr) & I(3)`.
-   Python applies `&` before the comparison. The parentheses in the cell that compares two matrices are not needed, but they help to read it.
-   The procedure `elim` of the Notebook for Section 4.2 has a variant for the rows. Its number is 4 and it is added to the others: `A.elim(4)` reduces to pre-echelon form with row operations and `A.elim(24)` reduces to row echelon form.



## What you should know at the end



-   An operation written on the left of the matrix operates on its rows.
-   ${}_{\boldsymbol{\tau}}\Mat{A}=\big({}_{\boldsymbol{\tau}}\Mat{I}\big)\Mat{A}$: operating on the rows of a matrix is multiplying it on the left by an elementary matrix.
-   ${}_{\boldsymbol{\tau}}\Mat{I}={\big({\Mat{I}}_{\boldsymbol{\tau}}\big)}^\T$: the elementary matrix for the rows is the transpose of the elementary matrix for the columns.
-   In a sequence of row operations, the one closest to the matrix acts first.
-   Transposing a sequence is reversing its order. When you transpose a matrix obtained with a sequence, the sequence changes sides and is transposed.
-   Every matrix can be reduced to row echelon form: you reduce its transpose to echelon form and transpose the result.
