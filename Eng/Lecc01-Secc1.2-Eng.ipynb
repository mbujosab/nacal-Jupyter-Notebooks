### Order of Operations and the Need for Parentheses



Now try removing the parentheses from the previous expression and run the code
(*remove the hash sign `#` and run the following code to see it*).



```python
# a|3 + b|3
```

Python will throw an error because it performs the addition `+` before the selection `|`: therefore, the first thing it tries is to add the number `3` and the vector $\Vect{b}$, but the addition of numbers and vectors is not defined, resulting in an error. Thus, it is necessary to use parentheses to indicate the correct order of operations. This allows you to calculate first what is inside the parentheses and then perform the addition of the two results.

Note that when we humans read $\;\eleVR{a}{3}+\eleVR{b}{3},\;$ we mentally interpret it as $\;\eleVRPE{a}{3}+\eleVRPE{b}{3};\;$, but Python requires you to clearly indicate the order of operations with parentheses.

When expressions are complex, even humans need parentheses for proper interpretation. Whenever you write a mathematical expression and have doubts, it is better to use parentheses. Remember that Python executes operations from left to right, but with strict precedence rules: addition and multiplication are performed before selection. Therefore, parentheses are usually necessary.
