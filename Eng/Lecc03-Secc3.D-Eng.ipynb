## The trace of a matrix



Read Definition 3.7, at the beginning of [Section 3.D](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph4). The trace of a matrix is the sum of the components of its main diagonal. In NAcAL `.diag()` gives the diagonal and `.traza()` gives the trace.



```python
Q = Matrix([ [1,2,3], [4,5,6], [7,8,10] ])
Q
```

```python
Q.diag()
```

```python
Q.traza()
```

The definition also holds for matrices that are not square. The following matrix has two rows and three columns. Say what its trace is before computing it.



```python
A = Matrix([ [1,2,0], [3,5,4] ])
A
```

## Properties of the trace



Read the exercise of [Section 3.D](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph4) and the two results that follow it. The book deduces some of these properties from the product seen as a sum of matrices, which is Section 3.C.1. The cells of this part also use these matrices.



```python
B = Matrix([ [2,1,-1], [0,3,5] ])
C = Matrix([ [1,0], [2,-1], [3,2] ])
K = Matrix([ [1,2], [0,3] ])
W = Matrix([ [2,0,1], [1,-1,0], [3,1,4] ])
```

First: the trace is a linear operator. The cell checks that the trace of a sum is the sum of the traces.



```python
(Q + W).traza() == Q.traza() + W.traza()
```

Check yourself that the trace of $5\Mat{Q}$ is 5 times the trace of $\Mat{Q}$. Write `(5*Q).traza()`. Without the parentheses, `.traza()` applies only to `Q`.



Second: a matrix and its transpose have the same trace. Check it with $\Mat{Q}$.



Third: $(\MatT{A})\Mat{A}$ and $\Mat{A}(\MatT{A})$ have the same trace. The two matrices have different orders.



```python
(~A) * A
```

```python
A * (~A)
```

```python
((~A)*A).traza() == (A*(~A)).traza()
```

Fourth: if $\Mat{A}$ and $\Mat{B}$ have the same order, the matrices $(\MatT{A})\Mat{B}$, $\Mat{B}(\MatT{A})$, $\Mat{A}(\MatT{B})$ and $(\MatT{B})\Mat{A}$ have the same trace. The cell computes the first one. Compute the other three.



```python
((~A)*B).traza()
```

### The trace of a product



The book deduces that $\Mat{AC}$ and $\Mat{CA}$ have the same trace. Here $\Mat{AC}$ has order 2 and $\Mat{CA}$ has order 3.



```python
A * C
```

```python
C * A
```

The two matrices are very different. Compute their traces.



With three matrices the book proves that $\Mat{ACK}$, $\Mat{KAC}$ and $\Mat{CKA}$ have the same trace. The cell computes the first one. Compute the other two.



```python
(A*C*K).traza()
```

## Powers of a square matrix



Read [Section 3.F](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph6). $\Mat{Q}^m$ is the product of $\Mat{Q}$ with itself $m$ times. In NAcAL the power is written with `**`.



```python
Q**3 == Q*Q*Q
```

The notation satisfies the law of exponents: $(\Mat{Q}^m)(\Mat{Q}^n)=\Mat{Q}^{m+n}$.



```python
(Q**2) * (Q**3) == Q**5
```

Check the law with two other exponents.



Powers only exist for square matrices. $\Mat{A}$ has two rows and three columns, and the product $\Mat{AA}$ does not exist. The next cell gives an error.



```python
A**2
```

### Zeroth power



Read [Section 3.F.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.alph2.3.Alph6.1). So that the law of exponents also holds with the exponent zero, the book defines $\Mat{Q}^0=\Mat{I}$.



```python
Q**0
```

Check that $(\Mat{Q}^0)(\Mat{Q}^3)=\Mat{Q}^3$.



At the end of the section the book explains what happens with negative exponents. They need the inverse matrix, which is the subject of Lecture 5.



## What you should know at the end



-   The trace of a matrix is the sum of the components of its main diagonal.
-   The trace is a linear operator, and a matrix and its transpose have the same trace.
-   $\Mat{AC}$ and $\Mat{CA}$ have the same trace, even if they are matrices of different orders.
-   $\Mat{Q}^m$ is the product of $\Mat{Q}$ with itself $m$ times, and it only exists if $\Mat{Q}$ is square.
-   $(\Mat{Q}^m)(\Mat{Q}^n)=\Mat{Q}^{m+n}$, and by definition $\Mat{Q}^0=\Mat{I}$.
