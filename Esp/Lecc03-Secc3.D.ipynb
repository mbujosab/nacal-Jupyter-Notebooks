## La traza de una matriz



Lea la Definición 3.7, al comienzo de la [Sección 3.D](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph4). La traza de una matriz es la suma de las componentes de su diagonal principal. En NAcAL `.diag()` da la diagonal y `.traza()` da la traza.



```python
Q = Matrix([ [1,2,3], [4,5,6], [7,8,10] ])
Q
```

```python
Q.diag()
```

```python
Q.traza()
```

La definición vale también para matrices que no son cuadradas. La matriz siguiente tiene dos filas y tres columnas. Diga cuál es su traza antes de calcularla.



```python
A = Matrix([ [1,2,0], [3,5,4] ])
A
```

## Propiedades de la traza



Lea el ejercicio de la [Sección 3.D](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph4) y los dos resultados que lo siguen. El libro deduce algunas de estas propiedades del producto visto como suma de matrices, que es la Sección 3.C.1. Las celdas de este apartado usan además estas matrices.



```python
B = Matrix([ [2,1,-1], [0,3,5] ])
C = Matrix([ [1,0], [2,-1], [3,2] ])
K = Matrix([ [1,2], [0,3] ])
W = Matrix([ [2,0,1], [1,-1,0], [3,1,4] ])
```

Primera: la traza es un operador lineal. La celda comprueba que la traza de una suma es la suma de las trazas.



```python
(Q + W).traza() == Q.traza() + W.traza()
```

Compruebe usted que la traza de $5\Mat{Q}$ es 5 por la traza de $\Mat{Q}$. Escriba `(5*Q).traza()`. Sin los paréntesis, `.traza()` se aplica solo a `Q`.



Segunda: una matriz y su transpuesta tienen la misma traza. Compruébelo con $\Mat{Q}$.



Tercera: $(\MatT{A})\Mat{A}$ y $\Mat{A}(\MatT{A})$ tienen la misma traza. Las dos matrices son de distinto orden.



```python
(~A) * A
```

```python
A * (~A)
```

```python
((~A)*A).traza() == (A*(~A)).traza()
```

Cuarta: si $\Mat{A}$ y $\Mat{B}$ son del mismo orden, las matrices $(\MatT{A})\Mat{B}$, $\Mat{B}(\MatT{A})$, $\Mat{A}(\MatT{B})$ y $(\MatT{B})\Mat{A}$ tienen la misma traza. La celda calcula la primera. Calcule las otras tres.



```python
((~A)*B).traza()
```

### La traza de un producto



El libro deduce que $\Mat{AC}$ y $\Mat{CA}$ tienen la misma traza. Aquí $\Mat{AC}$ es de orden 2 y $\Mat{CA}$ es de orden 3.



```python
A * C
```

```python
C * A
```

Las dos matrices son muy distintas. Calcule sus trazas.



Con tres matrices el libro demuestra que $\Mat{ACK}$, $\Mat{KAC}$ y $\Mat{CKA}$ tienen la misma traza. La celda calcula la primera. Calcule las otras dos.



```python
(A*C*K).traza()
```

## Potencias de una matriz cuadrada



Lea la [Sección 3.F](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph6). $\Mat{Q}^m$ es el producto de $\Mat{Q}$ consigo misma $m$ veces. En NAcAL la potencia se escribe con `**`.



```python
Q**3 == Q*Q*Q
```

La notación cumple la ley de exponentes: $(\Mat{Q}^m)(\Mat{Q}^n)=\Mat{Q}^{m+n}$.



```python
(Q**2) * (Q**3) == Q**5
```

Compruebe la ley con otros dos exponentes.



Las potencias solo existen para matrices cuadradas. $\Mat{A}$ tiene dos filas y tres columnas, y el producto $\Mat{AA}$ no existe. La celda siguiente da un error.



```python
A**2
```

### Potencia cero



Lea la [Sección 3.F.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.alph2.3.Alph6.1). Para que la ley de exponentes valga también con el exponente cero, el libro define $\Mat{Q}^0=\Mat{I}$.



```python
Q**0
```

Compruebe que $(\Mat{Q}^0)(\Mat{Q}^3)=\Mat{Q}^3$.



El libro explica al final de la sección qué ocurre con los exponentes negativos. Hace falta la matriz inversa, que es el tema de la Lección 5.



## Qué debe saber al terminar



-   La traza de una matriz es la suma de las componentes de su diagonal principal.
-   La traza es un operador lineal, y una matriz y su transpuesta tienen la misma traza.
-   $\Mat{AC}$ y $\Mat{CA}$ tienen la misma traza, aunque sean matrices de distinto orden.
-   $\Mat{Q}^m$ es el producto de $\Mat{Q}$ consigo misma $m$ veces, y solo existe si $\Mat{Q}$ es cuadrada.
-   $(\Mat{Q}^m)(\Mat{Q}^n)=\Mat{Q}^{m+n}$, y por definición $\Mat{Q}^0=\Mat{I}$.
