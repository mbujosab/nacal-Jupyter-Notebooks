## Una misma matriz elemental por las filas o por las columnas



Lea el comienzo de la [Sección 4.B](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph2). Una misma matriz elemental de Tipo I se puede obtener transformando las columnas de la identidad o transformando sus filas. Las dos transformaciones no son la misma: tienen los índices intercambiados.



```python
I(3) & T( (5,2,3) )
```

```python
T( (5,3,2) ) & I(3)
```

Las dos celdas dan la misma matriz. La primera suma a la tercera columna 5 veces la segunda. La segunda suma a la segunda fila 5 veces la tercera.

## El espejo de una transformación elemental



El libro llama espejo de una transformación a la transformación que da el mismo resultado al actuar por el otro lado de la matriz identidad. En NAcAL se obtiene con `.espejo()`. El espejo de una transformación de Tipo I tiene los dos índices intercambiados.



```python
T( (5,2,3) ).espejo()
```

Las transformaciones de Tipo II y los intercambios son su propio espejo, porque sus matrices son simétricas.



```python
T( (4,2) ).espejo()
```

```python
T( {2,3} ).espejo()
```

Compruebe con la matriz identidad de orden 3 que el intercambio anterior da la misma matriz por las filas y por las columnas.



## El espejo de una secuencia



El espejo de una secuencia de transformaciones es la secuencia de los espejos, en el mismo orden. La secuencia siguiente es la del ejemplo del libro.



```python
Tr = T( (3,2,1) ) & T( (-2,3,2) ) & T( (7,1,3) ) & T( {2,3} ) & T( (20,1) )
Tr
```

```python
Tr.espejo()
```

Solo han cambiado los índices de las tres transformaciones de Tipo I. El libro demuestra que el espejo aplicado a las filas de la identidad da la misma matriz que la secuencia aplicada a sus columnas.



```python
I(3) & Tr
```

```python
Tr.espejo() & I(3)
```

### El orden en que actúan



Las dos secuencias tienen las transformaciones en el mismo orden, pero no actúan en el mismo orden. En las columnas actúa primero la transformación de la izquierda. En las filas actúa primero la más cercana a la matriz, que es la última de la secuencia.

Aplique los cinco pasos de la secuencia espejo a las filas de la matriz `M`, uno en cada celda. Empiece por el que actúa primero. La primera celda ya está escrita. Al final debe obtener la matriz de las dos celdas anteriores.



```python
M = I(3)
T( (20,1) ) & M
```

### El espejo no es la transpuesta



No confunda el espejo de una secuencia con su transpuesta, que es del Notebook de la Sección 4.A. El espejo cambia los índices y conserva el orden. La transpuesta conserva los índices e invierte el orden.



```python
~Tr
```

La secuencia transpuesta aplicada a las filas de la identidad da la transpuesta de la matriz `I(3) & Tr`. Compruébelo.



### Con una matriz cualquiera



Transformar las filas de una matriz es multiplicarla por la izquierda por la matriz que resulta de transformar las filas de la identidad. Para la secuencia espejo esa matriz es ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$. Por tanto, aplicar la secuencia espejo a las filas de $\Mat{A}$ es multiplicar $\Mat{A}$ por la izquierda por ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$.



```python
A = Matrix([ [1,2,0], [3,-1,4], [2,5,1] ])
Tr.espejo() & Matrix(A)
```

```python
( I(3) & Tr ) * A
```

Así la misma matriz ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ multiplica por la derecha cuando la secuencia se aplica a las columnas, y por la izquierda cuando su espejo se aplica a las filas. Compruebe lo primero.



## Matrices simétricas y transformación espejo



Lea el párrafo «Matrices simétricas y transformación espejo» de la [Sección 4.B](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph2). La matriz ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ es simétrica si y solo si el espejo de la secuencia y su transpuesta dan la misma matriz al aplicarlos a las filas de la identidad. La secuencia siguiente es la del ejemplo del libro.



```python
p = T( [ (9,2), (18,1), (-1,2,1), (fracc(1,9),1), (-1,1,2), (fracc(1,2),2) ], rpr='h' )
(p.espejo() & I(2)) == (~p & I(2))
```

```python
p.espejo() & I(2)
```

Las dos secuencias dan la misma matriz, pero no son la misma secuencia. Compare las tres.



```python
p
```

```python
p.espejo()
```

```python
~p
```

## El espejo de la inversa de una transformación



Lea el final de la [Sección 4.B](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph2). La inversa de una transformación elemental es la transformación que la deshace. En NAcAL se escribe con `**-1`. La Lección 5 trata de las matrices inversas.



```python
T( (3,1,2) )**-1
```

```python
T( (2,3) )**-1
```

La inversa de una secuencia deshace los pasos en el orden contrario. La Lección 5 lo demuestra.



```python
Tr**-1
```

Aplique a una copia de $\Mat{A}$ la secuencia `Tr` y después su inversa. Debe obtener de nuevo la matriz $\Mat{A}$.



El libro deduce que, si a las columnas de la identidad se les aplica una secuencia y a sus filas el espejo de la inversa, se obtiene de nuevo la identidad.



```python
(Tr**-1).espejo() & I(3) & Tr
```

La Lección 17 aplica esas dos secuencias a una matriz cuadrada cualquiera: la secuencia a sus columnas y el espejo de la inversa a sus filas. La matriz que se obtiene se llama semejante a la primera. La celda siguiente lo hace con una secuencia corta.



```python
S = T( [ (2,1,2), (-1,3,1) ] )
(S**-1).espejo() & Matrix(A) & S
```

Escriba la secuencia `(S**-1).espejo()` y diga qué hace cada uno de sus pasos con las filas.



## Detalles de Python



-   `.espejo()` y `**-1` se aplican a una transformación o a una secuencia y dan otra secuencia. No cambian la secuencia original.
-   Los paréntesis de `(Tr**-1).espejo()` son necesarios. Sin ellos Python da un error.
-   Python calcula `**` antes que `&`. Por eso `Tr**-1` no necesita paréntesis en `Matrix(A) & Tr & Tr**-1`.
-   En la celda `(Tr**-1).espejo() & I(3) & Tr` no hay copia, porque `I(3)` crea una matriz identidad nueva cada vez.



## Qué debe saber al terminar



-   El espejo de una transformación es la transformación que da la misma matriz al actuar por el otro lado de la identidad.
-   El espejo de una transformación de Tipo I tiene los índices intercambiados. Las de Tipo II y los intercambios son su propio espejo.
-   El espejo de una secuencia es la secuencia de los espejos, en el mismo orden. Aplicado a las filas de la identidad da la misma matriz que la secuencia aplicada a las columnas.
-   El espejo cambia los índices y conserva el orden. La transpuesta conserva los índices e invierte el orden.
-   La matriz ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ es simétrica si y solo si el espejo de la secuencia y su transpuesta dan la misma matriz al aplicarlos a las filas de la identidad.
-   Si a las columnas de la identidad se les aplica una secuencia y a sus filas el espejo de su inversa, se obtiene de nuevo la identidad. La Lección 17 usa esas dos secuencias para obtener matrices semejantes.
