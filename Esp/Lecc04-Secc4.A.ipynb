## Transformar las filas



Lea el comienzo de la [Sección 4.A](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph1). Para transformar las filas de una matriz el libro escribe la transformación a la izquierda: ${}_{\boldsymbol{\tau}}\Mat{A}$. En NAcAL también se escribe a la izquierda. Las celdas de este Notebook usan estas dos matrices.



```python
A = Matrix([ [1,2,0], [3,-1,4], [2,5,1] ])
B = Matrix([ [2,1], [0,-1], [1,3] ])
A
```

La celda siguiente suma a la tercera fila 2 veces la primera. Como en el Notebook de la Sección 4.1, la transformación se aplica a una copia. Si cambia una de las dos matrices sin querer, ejecute de nuevo la celda que las define.



```python
T( (2,1,3) ) & Matrix(A)
```

La misma transformación, escrita a la derecha, suma a la tercera columna 2 veces la primera.



```python
Matrix(A) & T( (2,1,3) )
```

El libro llega a las transformaciones de las filas con la transpuesta. Se transforman las columnas de $\MatT{A}$ y se transpone el resultado.



```python
~( (~A) & T( (2,1,3) ) )
```

Es la misma matriz que ha dado la primera de las tres celdas.



### Transformar las filas es multiplicar por la izquierda



El libro demuestra que ${}_{\boldsymbol{\tau}}\Mat{A}=\big({}_{\boldsymbol{\tau}}\Mat{I}\big)\Mat{A}$. Transformar las filas de $\Mat{A}$ da lo mismo que multiplicar $\Mat{A}$ por la izquierda por la matriz que resulta de transformar las filas de la identidad.



```python
T( (2,1,3) ) & I(3)
```

```python
( T( (2,1,3) ) & I(3) ) * A
```

Compruebe la igualdad con la matriz $\Mat{B}$ y la transformación que multiplica por 5 la segunda fila. ¿De qué orden debe ser la matriz identidad?



### La matriz elemental de las filas es la transpuesta de la de las columnas



El libro demuestra también que ${}_{\boldsymbol{\tau}}\Mat{I}={\big({\Mat{I}}_{\boldsymbol{\tau}}\big)}^\T$.



```python
T( (2,1,3) ) & I(3)
```

```python
I(3) & T( (2,1,3) )
```

Mire dónde está el 2 en cada matriz. Al transformar las columnas está en la fila 1 y la columna 3, como en el Notebook de la Sección 4.1. Al transformar las filas está en la fila 3 y la columna 1.

Aplique una transformación de Tipo II a las filas de la identidad y después a sus columnas. ¿Por qué se obtiene la misma matriz?



## Secuencias de transformaciones de las filas



En una secuencia escrita a la izquierda de la matriz actúa primero la transformación más cercana a la matriz. La celda siguiente primero multiplica por 2 la segunda fila y después intercambia las dos primeras filas.



```python
T( {1,2} ) & T( (2,2) ) & Matrix(A)
```

Escriba las mismas dos transformaciones en el otro orden y diga, antes de ejecutar, qué fila quedará multiplicada por 2.



### La transpuesta de una secuencia



El libro llama transpuesta de una secuencia a la secuencia con las mismas transformaciones en el orden contrario. En NAcAL se escribe con `~`. La secuencia siguiente es la del ejemplo del libro.



```python
Tr = T( [ (66,1,2), {3,2}, (4,1) ], rpr='h' )
Tr
```

```python
~Tr
```

La secuencia aplicada a las columnas de la identidad y su transpuesta aplicada a las filas dan dos matrices. Cada una es la transpuesta de la otra.



```python
I(3) & Tr
```

```python
~Tr & I(3)
```

Aplique ahora la secuencia sin transponer a las filas de la identidad. No se obtiene ninguna de las dos matrices anteriores.



Con una matriz cualquiera pasa lo mismo que con un producto de matrices. Al transponer, la secuencia cambia de lado y se transpone: ${\big({\Mat{A}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}\big)}^\T={}_{(\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k)^\T}\big(\MatT{A}\big)$.



```python
~( Matrix(A) & Tr )
```

```python
~Tr & (~A)
```

### La notación sigue siendo asociativa



El libro recuadra que $\big({}_{\boldsymbol{\tau}}\Mat{A}\big)\Mat{B}={}_{\boldsymbol{\tau}}\big(\Mat{A}\Mat{B}\big)$. Compruébelo con la secuencia `Tr` y las matrices $\Mat{A}$ y $\Mat{B}$.



## Matrices simétricas



Lea el párrafo «Matrices simétricas» de la [Sección 4.A](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph3.4.Alph1). La matriz ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ es simétrica si y solo si coincide con la matriz que resulta de aplicar la secuencia transpuesta a las filas de la identidad. La secuencia siguiente es la del ejemplo del libro.



```python
p = T( [ (9,2), (18,1), (-1,2,1), (fracc(1,9),1), (-1,1,2), (fracc(1,2),2) ], rpr='h' )
p
```

```python
(~p & I(2)) == (I(2) & p)
```

```python
I(2) & p
```

## Matriz escalonada por filas



Lea la [Sección 4.A.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.alph3.4.Alph1.1). Una matriz es escalonada por filas si su transpuesta es escalonada. El pivote de una fila no nula es su primera componente no nula. `ElimGF` escalona con transformaciones de las filas. La matriz siguiente es la del ejemplo del libro.



```python
G = Matrix([ [0,2,2], [1,-2,-8], [3,6,-6] ])
ElimGF(G, 1)
```

Las transformaciones de las filas se escriben debajo de las flechas. Cuando hay varias en un paso, actúa primero la de abajo.

El libro escalona por filas así: escalona la transpuesta y transpone el resultado. Compruébelo con `ElimG`, que es del [Notebook de la Sección 4.2](Lecc04-Secc4.2.ipynb).



Escalone ahora por filas la matriz $\Mat{A}$ en papel. Después aplique sus transformaciones a la copia `M`. Recuerde que se escriben a la izquierda.



```python
M = Matrix(A)
```

```python
ElimGF(A, 1)
```

## Detalles de Python



-   Con `rpr='h'` la secuencia se muestra en horizontal. Sin ese argumento se muestra en vertical.
-   Python aplica `~` antes que `&`. Por eso `~Tr & I(3)` es lo mismo que `(~Tr) & I(3)`.
-   Python aplica `&` antes que la comparación. Los paréntesis de la celda que compara dos matrices no son necesarios, pero ayudan a leerla.
-   El procedimiento `elim` del Notebook de la Sección 4.2 tiene una variante para las filas. Su número es el 4 y se suma a los demás: `A.elim(4)` pre-escalona por filas y `A.elim(24)` escalona por filas.



## Qué debe saber al terminar



-   Una transformación escrita a la izquierda de la matriz transforma sus filas.
-   ${}_{\boldsymbol{\tau}}\Mat{A}=\big({}_{\boldsymbol{\tau}}\Mat{I}\big)\Mat{A}$: transformar las filas de una matriz es multiplicarla por la izquierda por una matriz elemental.
-   ${}_{\boldsymbol{\tau}}\Mat{I}={\big({\Mat{I}}_{\boldsymbol{\tau}}\big)}^\T$: la matriz elemental de las filas es la transpuesta de la matriz elemental de las columnas.
-   En una secuencia de transformaciones de las filas actúa primero la más cercana a la matriz.
-   Transponer una secuencia es invertir su orden. Al transponer una matriz transformada, la secuencia cambia de lado y se transpone.
-   Toda matriz se puede escalonar por filas: se escalona su transpuesta y se transpone el resultado.
