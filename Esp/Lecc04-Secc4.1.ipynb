## Transformaciones elementales de Tipo I



Lea el comienzo de la Lección 4 y la [Sección 4.1.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.1). Una transformación elemental de Tipo I suma a un vector un múltiplo de otro vector. En NAcAL se escribe con `T` y tres números entre paréntesis. La siguiente suma 8 veces el segundo vector al tercero.



```python
T( (8,2,3) )
```

El primer número es el que multiplica. El último es el índice del vector que cambia.

El operador `&` aplica la transformación a las columnas de la matriz que tiene a su izquierda. Si esa matriz es la identidad, el resultado es una matriz elemental.



```python
I(3) & T( (8,2,3) )
```

Mire dónde está el 8: en la fila 2 y la columna 3. Son los dos índices de la transformación, en el mismo orden.

Escriba la matriz elemental de orden 4 que resulta de sumar $-7$ veces la primera columna a la tercera. Diga antes en qué fila y en qué columna estará el $-7$.



Las celdas de este Notebook usan estas tres matrices.



```python
A = Matrix([ [1,2,0], [3,-1,4] ])
B = Matrix([ [2,1], [0,-1], [1,3] ])
C = Matrix([ [0,1,5], [7,0,4] ])
A
```

La celda siguiente suma 2 veces la primera columna a la tercera. La transformación se aplica a `Matrix(A)`, que es una copia de $\Mat{A}$.



```python
Matrix(A) & T( (2,1,3) )
```

Compare con $\Mat{A}$. Solo ha cambiado la tercera columna.



### La transformación cambia la matriz



El operador `&` no crea una matriz nueva: cambia la matriz que tiene a su izquierda. Por eso la celda anterior usa una copia. Las tres celdas siguientes lo muestran con otra copia, llamada `D`.



```python
D = Matrix(A)
D & T( (2,1,3) )
```

```python
D
```

```python
A
```

La matriz `D` ha cambiado y la matriz `A` no. En este Notebook las transformaciones se aplican casi siempre a una copia. Así las tres matrices del principio sirven para todas las celdas. Si cambia una de estas matrices sin querer, ejecute de nuevo la celda que las define.



## Transformaciones elementales de Tipo II



Lea la [Sección 4.1.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.2). Una transformación elemental de Tipo II multiplica un vector por un número distinto de cero. En NAcAL se escribe con dos números: el número que multiplica y el índice del vector. La siguiente multiplica por 5 el primer vector.



```python
T( (5,1) )
```

```python
I(3) & T( (5,1) )
```

La matriz elemental es como la identidad, salvo por la primera componente de la diagonal.

Multiplique por $-10$ la segunda columna de una copia de $\Mat{A}$.



### Lo que no es una transformación elemental



Multiplicar una columna por cero no es una transformación elemental. Restar una columna a sí misma tampoco lo es. NAcAL no las admite.



```python
T( (0,2) )
```

NAcAL da un error. Pruebe con `T( (-1,2,2) )`, que restaría el segundo vector a sí mismo.



## Transformar las columnas es multiplicar por una matriz elemental



Lea la [Sección 4.1.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.3). El libro demuestra que ${\Mat{A}}_{\boldsymbol{\tau}}=\Mat{A}\big({\Mat{I}}_{\boldsymbol{\tau}}\big)$. Aplicar una transformación elemental a las columnas de $\Mat{A}$ da lo mismo que multiplicar $\Mat{A}$ por la matriz elemental de esa transformación.



```python
Matrix(A) & T( (2,1,3) )
```

```python
A * ( I(3) & T( (2,1,3) ) )
```

Las dos celdas dan la misma matriz. La matriz identidad es de orden 3 porque $\Mat{A}$ tiene tres columnas.

Compruebe la igualdad con la transformación de Tipo II que multiplica por 5 la segunda columna.



### Una nueva regla de reescritura



De esa igualdad el libro deduce que ${\big(\Mat{A}\Mat{B}\big)}_{\boldsymbol{\tau}}=\Mat{A}\big({\Mat{B}}_{\boldsymbol{\tau}}\big)$. La transformación de las celdas siguientes resta de la segunda columna 3 veces la primera.



```python
(A*B) & T( (-3,1,2) )
```

```python
A * ( Matrix(B) & T( (-3,1,2) ) )
```

Las dos celdas dan la misma matriz. Por eso el libro escribe ${\Mat{A}\Mat{B}}_{\boldsymbol{\tau}}$ sin paréntesis. Calcule $\Mat{AB}$ y diga qué columna ha cambiado.



## Las matrices elementales y su notación



Lea la [Sección 4.1.4](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.4). La notación de una transformación no dice de qué orden es su matriz elemental. La misma transformación da una matriz elemental de cada orden.



```python
I(2) & T( (-7,1,2) )
```

```python
I(4) & T( (-7,1,2) )
```

La matriz identidad es una matriz elemental de los dos tipos. Se obtiene al sumar a una columna cero veces otra. También se obtiene al multiplicar una columna por 1. NAcAL admite las dos transformaciones.



```python
I(3) & T( (0,1,2) )
```

```python
I(3) & T( (1,2) )
```

La transpuesta de una matriz elemental es otra matriz elemental del mismo tipo. En las de Tipo I se intercambian los dos índices.



```python
~( I(3) & T( (8,2,3) ) )
```

```python
I(3) & T( (8,3,2) )
```

Compruebe que una matriz elemental de Tipo II es igual a su transpuesta.

## Secuencias de transformaciones elementales



Lea la [Sección 4.1.5](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.5). Con `&` también se encadenan transformaciones para formar una secuencia. Se aplican de izquierda a derecha. La celda siguiente es el ejemplo del libro. Primero multiplica por 3 la segunda columna de $\Mat{C}$. Después suma a la segunda columna 2 veces la primera.



```python
Matrix(C) & T( (3,2) ) & T( (2,1,2) )
```

Una secuencia se puede guardar en una variable. También se puede escribir como una lista dentro de `T`.



```python
Tr = T( [ (3,2), (2,1,2) ] )
Tr
```

```python
Matrix(C) & Tr
```

El orden importa. Aplique a una copia de $\Mat{C}$ las mismas dos transformaciones en el orden contrario. ¿Qué componente es distinta?



### Una secuencia es un producto de matrices elementales



La matriz ${\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}$ es el producto de las matrices elementales de la secuencia, en el mismo orden.



```python
I(3) & Tr
```

```python
( I(3) & T( (3,2) ) ) * ( I(3) & T( (2,1,2) ) )
```

Las dos celdas dan la misma matriz. No es una matriz elemental: se distingue de la identidad en dos componentes.

El libro deduce que ${\Mat{A}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}=\Mat{A}\big({\Mat{I}}_{\boldsymbol{\tau}_1\cdots\boldsymbol{\tau}_k}\big)$. Multiplique $\Mat{C}$ por la matriz anterior y compare con la matriz que ha obtenido al aplicar la secuencia a $\Mat{C}$.



### Una secuencia es un operador lineal



Un ejercicio de la Sección 4.1.5 pide demostrar que aplicar una secuencia de transformaciones a las columnas es un operador lineal. La demostración es un trabajo para el papel. Aquí se comprueba en un caso. Primero con el producto por un escalar.



```python
(5*C) & Tr
```

```python
5 * ( Matrix(C) & Tr )
```

Compruebe ahora que transformar la suma $\Mat{A}+\Mat{C}$ da lo mismo que sumar las dos matrices transformadas.



## Intercambios



Lea la [Sección 4.1.6](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.6). El intercambio de dos vectores se escribe en NAcAL con los dos índices entre llaves.



```python
T( {1,3} )
```

```python
Matrix(A) & T( {1,3} )
```

Una matriz intercambio resulta de intercambiar dos columnas de la matriz identidad.



```python
I(3) & T( {1,3} )
```

El libro demuestra que toda matriz intercambio es simétrica y que su cuadrado es la matriz identidad. Compruebe las dos cosas con la matriz anterior. Para la primera use `.es_simetrica()`.



### El intercambio es una secuencia de transformaciones elementales



El libro no cuenta el intercambio como transformación elemental, porque se consigue con tres transformaciones de Tipo I y una de Tipo II. Haga en papel el primer apartado del ejercicio de la Sección 4.1.6. La celda siguiente aplica a la matriz identidad la secuencia de la solución del libro.



```python
I(3) & T( [ (1,1,2), (-1,2,1), (1,1,2), (-1,1) ] )
```

Cada paso se ve mejor en una matriz cuyas columnas sean distintas. La celda siguiente da el primer paso con una copia de $\Mat{A}$.



```python
D = Matrix(A)
D & T( (1,1,2) )
```

Dé los otros tres pasos, uno en cada celda. Cada celda cambia `D`, así que no hay que copiarla de nuevo. Al final las dos primeras columnas de $\Mat{A}$ están intercambiadas. Si ejecuta una celda dos veces, la transformación se aplica dos veces. En ese caso ejecute de nuevo la celda que define `D` y empiece otra vez.



## Permutaciones



Lea la [Sección 4.1.7](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.1.7). Una secuencia de intercambios reordena las columnas. Aplicada a la matriz identidad da una matriz permutación.



```python
I(4) & T( [ {1,3}, {4,2}, {3,2} ] )
```

Las matrices intercambio son simétricas. Las matrices permutación no lo son en general. La siguiente es la del ejemplo del libro.



```python
P = I(3) & T( [ {1,3}, {1,2} ] )
P
```

```python
P.es_simetrica()
```

## Detalles de Python



-   `T` recibe un solo argumento. Por eso hay dos pares de paréntesis en `T( (8,2,3) )`. Tres números entre paréntesis son una transformación de Tipo I. Dos números son una de Tipo II. Dos índices entre llaves son un intercambio.
-   Una lista de transformaciones dentro de `T` es una secuencia. `T( [ (3,2), (2,1,2) ] )` es lo mismo que `T( (3,2) ) & T( (2,1,2) )`.
-   El operador `&` cambia la matriz que tiene a su izquierda. `Matrix(A)` crea una copia de la matriz `A`.
-   `A*B` y `~A` crean matrices nuevas. No hace falta copiarlas antes de transformarlas.



## Qué debe saber al terminar



-   Hay dos tipos de transformaciones elementales. Las de Tipo I suman a un vector un múltiplo de otro vector. Las de Tipo II multiplican un vector por un número distinto de cero.
-   Una matriz elemental resulta de aplicar una sola transformación elemental a las columnas de la matriz identidad. Su transpuesta es otra matriz elemental del mismo tipo.
-   ${\Mat{A}}_{\boldsymbol{\tau}}=\Mat{A}\big({\Mat{I}}_{\boldsymbol{\tau}}\big)$: transformar las columnas de una matriz es multiplicarla por la derecha por una matriz elemental. Por eso ${\big(\Mat{A}\Mat{B}\big)}_{\boldsymbol{\tau}}=\Mat{A}\big({\Mat{B}}_{\boldsymbol{\tau}}\big)$.
-   Aplicar una secuencia de transformaciones es multiplicar por un producto de matrices elementales. El orden de las transformaciones importa.
-   Aplicar una secuencia de transformaciones a las columnas es un operador lineal.
-   El intercambio es una secuencia de transformaciones de Tipo I y una de Tipo II. Las matrices intercambio son simétricas y su cuadrado es la identidad. Las matrices permutación no son simétricas en general.
-   En NAcAL el operador `&` cambia la matriz sobre la que actúa.
