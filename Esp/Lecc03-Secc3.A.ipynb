## Matrices triangulares



Lea las tres definiciones del comienzo de la [Sección 3.A](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph1). Una matriz es triangular superior si son nulas todas las componentes que están por debajo de la diagonal principal. Es triangular inferior si son nulas todas las que están por encima.



```python
T = Matrix([ [2,1,3], [0,4,5], [0,0,6] ])
T
```

```python
T.es_triangularSup()
```

La transpuesta de una matriz triangular superior es triangular inferior. Compruébelo con `.es_triangularInf()`.



Una matriz triangular no tiene que ser cuadrada. La siguiente tiene dos filas y tres columnas.



```python
R = Matrix([ [1,2,3], [0,4,5] ])
R.es_triangularSup()
```

El libro dice que una matriz nula es triangular superior y también triangular inferior. En NAcAL la matriz nula de orden 2 por 3 es `M0(2,3)`. Compruebe las dos cosas.



Una matriz triangular es unitaria si todas las componentes de su diagonal son iguales a uno. En NAcAL `.diag()` da la diagonal. La matriz $\Mat{T}$ no es unitaria.



```python
T.diag()
```

## Producto de dos matrices triangulares superiores



Lea el ejercicio de la [Sección 3.A](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph1), que enuncia las proposiciones 3.A.1 a 3.A.3. Para ver qué ocurre con cualquier matriz, las dos siguientes tienen letras en lugar de números.



```python
a, b, c, d, e, f = sympy.symbols('a b c d e f')
p, q, r, s, t, u = sympy.symbols('p q r s t u')
U = Matrix([ [a,b,c], [0,d,e], [0,0,f] ])
V = Matrix([ [p,q,r], [0,s,t], [0,0,u] ])
U
```

```python
V
```

Proposición 3.A.1: el producto de dos matrices triangulares superiores es triangular superior.



```python
U * V
```

Mire las componentes que están por debajo de la diagonal: son nulas. Mire después la diagonal. ¿Qué relación ve entre la diagonal de $\Mat{UV}$ y las diagonales de $\Mat{U}$ y de $\Mat{V}$?



### Cuando una de las dos es unitaria



Las proposiciones 3.A.2 y 3.A.3 piden matrices cuadradas y del mismo orden. La matriz siguiente es $\Mat{U}$ con unos en la diagonal.



```python
U1 = Matrix([ [1,b,c], [0,1,e], [0,0,1] ])
U1
```

Proposición 3.A.2: si la primera matriz es triangular superior unitaria y la segunda es triangular superior, el producto es triangular superior y mantiene la diagonal de la segunda.



```python
U1 * V
```

La diagonal del producto es la de $\Mat{V}$. Proposición 3.A.3: si la primera es triangular superior y la segunda es triangular superior unitaria, el producto mantiene la diagonal de la primera. Compruébela con $\Mat{V}$ y $\Mat{U1}$.



El Corolario 3.A.4 dice que el producto de dos matrices triangulares superiores unitarias, cuadradas y del mismo orden, es también triangular superior unitaria. Compruébelo con el cuadrado de $\Mat{U1}$, que es `U1*U1`.



## Matrices triangulares inferiores



El libro dice que los mismos resultados son ciertos con matrices triangulares inferiores, y que basta transponer para demostrarlo. Las transpuestas de $\Mat{U}$ y de $\Mat{V}$ son triangulares inferiores. Multiplíquelas y mire dónde están los ceros.



Compare su resultado con la transpuesta de $\Mat{VU}$. Recuerde la Proposición 3.2.2: al transponer un producto cambia el orden de los factores.



## Qué debe saber al terminar



-   Una matriz es triangular superior si son nulas las componentes que están por debajo de la diagonal principal. Es triangular inferior si su transpuesta es triangular superior.
-   Una matriz triangular es unitaria si su diagonal está formada por unos.
-   El producto de dos matrices triangulares superiores es triangular superior. Lo mismo ocurre con las inferiores.
-   Si las dos son cuadradas y del mismo orden y una es unitaria, el producto mantiene la diagonal de la otra.
-   El producto de dos triangulares superiores unitarias, cuadradas y del mismo orden, es triangular superior unitaria.
