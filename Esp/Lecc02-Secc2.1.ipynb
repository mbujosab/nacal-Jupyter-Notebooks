## El producto punto



Lea la definición de producto punto al comienzo de la [Sección 2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.1). El producto punto de dos vectores de $\R[n]$ es un número: la suma de los productos de sus componentes. En NAcAL se escribe con el símbolo `*`.



```python
a = Vector([1, 2, 3])
b = Vector([1, -1, 2])
a * b
```

Calcule el mismo número con las componentes de los dos vectores. Multiplique las primeras componentes, las segundas y las terceras, y sume los tres productos. El primer producto es `(a|1)*(b|1)`.



Los dos vectores deben tener el mismo número de componentes. Ejecute la celda siguiente y lea el mensaje de error.



```python
a * Vector([1, 2])
```

El mensaje está en la última línea: `ValueError: Sistema y Vector incompatibles`. Las líneas anteriores indican en qué punto del código de NAcAL se ha detectado el error. No hace falta leerlas. El mensaje dice que los dos vectores no se pueden multiplicar: uno tiene tres componentes y el otro dos.



## Propiedades del producto punto



El primer ejercicio de la [Sección 2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.1) pide demostrar cuatro propiedades. Cada celda siguiente compara los dos lados de una igualdad con los vectores $\Vect{a}$, $\Vect{b}$ y $\Vect{c}$.



```python
c = Vector([2, 1, -1])
```

Simetría: $\Vect{a}\cdot\Vect{b}=\Vect{b}\cdot\Vect{a}$.



```python
a * b == b * a
```

Lineal en el primer argumento. Asociativo respecto al producto por escalares: $(\lambda\Vect{a})\cdot\Vect{b}=\lambda(\Vect{a}\cdot\Vect{b})$.



```python
(3*a) * b == 3 * (a*b)
```

Distributivo respecto a la suma: $(\Vect{a}+\Vect{b})\cdot\Vect{c}=\Vect{a}\cdot\Vect{c}+\Vect{b}\cdot\Vect{c}$. Escríbalo usted.



No negativo: $\Vect{x}\cdot\Vect{x}\geq0$. Calcule el producto punto de $\Vect{b}$ por sí mismo. La segunda componente de $\Vect{b}$ es negativa. Calcule también `(b|2)*(b|2)`: es el sumando que aporta esa componente.



Definido: $\Vect{x}\cdot\Vect{x}=0$ si y solo si $\Vect{x}$ es el vector nulo. Calcule el producto punto del vector nulo de $\R[3]$ por sí mismo. El vector nulo es `V0(3)`. Pruebe después con un vector que tenga una sola componente distinta de cero. ¿Puede dar 0?



Estas celdas comprueban cada propiedad para estos vectores. No la demuestran para todos los vectores. La demostración está en el libro.



## El producto punto también es lineal en el segundo argumento



El libro lo deduce de la simetría. Compruebe las dos igualdades: $\Vect{a}\cdot(\lambda\Vect{b})=\lambda(\Vect{a}\cdot\Vect{b})$ y $\Vect{a}\cdot(\Vect{b}+\Vect{c})=\Vect{a}\cdot\Vect{b}+\Vect{a}\cdot\Vect{c}$. Use $\lambda=3$.



Como el producto punto es lineal en los dos argumentos, el escalar puede multiplicar a cualquiera de los dos vectores.



```python
a * (3*b) == (3*a) * b
```

## Producto punto por una fila o una columna de la matriz identidad



Lea el caso especial que cierra la [Sección 2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.1). La matriz identidad de orden 3 es `I(3)`.



```python
(1|I(3)) * b
```

```python
b * (I(3)|2)
```

La primera celda da la primera componente de $\Vect{b}$. La segunda da la segunda componente. Obtenga ahora la tercera componente de $\Vect{b}$ con un producto punto.



## El producto punto opera con dos vectores



Lea en la [Sección 2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.1) el párrafo anterior a las propiedades del producto punto. Dice que $(\Vect{a}\cdot\Vect{b})\Vect{c}$ y $\Vect{a}(\Vect{b}\cdot\Vect{c})$ son en general vectores distintos. Ejecute las dos celdas y compare los resultados.



```python
a * b * c
```

```python
a * (b * c)
```

Las dos celdas dan vectores distintos. En la primera, Python calcula antes `a*b`, que es un número, y multiplica ese número por $\Vect{c}$. En la segunda calcula antes `b*c` y multiplica $\Vect{a}$ por ese número. Ninguna de las dos es un producto punto de tres vectores. Ese producto no existe. Escriba siempre los paréntesis.



## Qué debe saber al terminar



-   El producto punto de dos vectores de $\R[n]$ es un número.
-   El producto punto es simétrico, lineal en sus dos argumentos, no negativo y definido.
-   El producto punto de un vector por la fila o la columna $i$-ésima de la matriz identidad es la componente $i$-ésima del vector.
-   El producto punto opera con dos vectores. $(\Vect{a}\cdot\Vect{b})\Vect{c}$ y $\Vect{a}(\Vect{b}\cdot\Vect{c})$ son en general vectores distintos.
-   Comprobar una igualdad con un ejemplo no es demostrarla.
