## La función `dibuja`



NAcAL no dibuja. La celda siguiente define una función, `dibuja`, que representa vectores de $\R[2]$ como flechas. Usa la librería [Matplotlib](https://matplotlib.org/). No hace falta entender su código. Ejecute la celda y pase al apartado siguiente.



```python
import matplotlib.pyplot as plt

def dibuja(*vectores, encadenados=False, nombres=None):
    """Dibuja vectores de R^2 como flechas.

    dibuja(a, b, c)   Todas las flechas salen del origen.

    dibuja(a, b, c, encadenados=True)
        Cada flecha sale de la punta de la anterior. Se dibuja
        además, en negro, la suma de todos los vectores.

    dibuja(a, b, nombres=['a', 'b'])
        Cambia el texto de la leyenda. Sin este argumento la
        leyenda muestra las componentes de cada vector."""
    for v in vectores:
        if not isinstance(v, Vector) or v.n != 2:
            raise ValueError('dibuja solo admite vectores de R^2. '
                             'Las opciones se escriben con su nombre: encadenados=True')
    colores = ['tab:blue', 'tab:red', 'tab:green', 'tab:orange', 'tab:purple', 'tab:brown']
    fig, ax = plt.subplots()

    def flecha(inicio, fin, color, nombre):
        ax.annotate('', xy=fin, xytext=inicio,
                    arrowprops=dict(arrowstyle='->', color=color, lw=2))
        ax.plot([], [], color=color, lw=2, label=nombre)

    def texto(x, y):
        return '(' + str(x) + ', ' + str(y) + ',)'

    puntos = [(0, 0)]
    inicio = (0, 0)
    suma = (0, 0)
    for k, v in enumerate(vectores):
        fin = (inicio[0] + float(1|v), inicio[1] + float(2|v))
        nombre = nombres[k] if nombres else texto(1|v, 2|v)
        flecha(inicio, fin, colores[k % len(colores)], nombre)
        puntos.append(fin)
        suma = (suma[0] + (1|v), suma[1] + (2|v))
        if encadenados:
            inicio = fin
    if encadenados:
        flecha((0, 0), inicio, 'black', 'suma: ' + texto(suma[0], suma[1]))

    xs = [p[0] for p in puntos]
    ys = [p[1] for p in puntos]
    ax.set_xlim(min(xs) - 1, max(xs) + 1)
    ax.set_ylim(min(ys) - 1, max(ys) + 1)
    ax.set_aspect('equal')
    ax.grid(True)
    ax.axhline(0, color='gray', lw=1)
    ax.axvline(0, color='gray', lw=1)
    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1))
    plt.show()
```

## Vectores desde el origen



El eje horizontal es la primera componente y el eje vertical es la segunda. Cada vector es una flecha que sale del origen y termina en el punto que indican sus componentes.



```python
a = Vector([1, 2])
b = Vector([3, -1])
dibuja(a, b)
```

La leyenda muestra las componentes de cada vector. El argumento `nombres` cambia ese texto.



```python
dibuja(a, b, nombres=['a', 'b'])
```

Dibuje ahora $2\Vect{a}$, $\Vect{a}$ y $-\Vect{a}$, en ese orden, en una misma figura. Con ese orden la flecha de $\Vect{a}$ queda encima de la de $2\Vect{a}$ y se ve. Mire qué tienen en común las tres flechas.



Dibuje $\Vect{a}$, $\Vect{b}$ y $\Vect{a}+\Vect{b}$ en una misma figura.



## La suma: una flecha a continuación de otra



Con `encadenados=True` cada flecha sale de la punta de la anterior. La flecha negra es la suma. Sale del origen y termina donde termina la última flecha.



```python
dibuja(a, b, encadenados=True)
```

Cambie el orden de los dos vectores y repita el dibujo. Mire dónde termina la flecha negra en los dos casos. ¿Qué propiedad de la suma está viendo?



Restar un vector es sumar su opuesto. Dibuje así $\Vect{a}-\Vect{b}$: encadene $\Vect{a}$ y $-\Vect{b}$.



## Combinaciones lineales



Una combinación lineal de $\Vect{a}$ y $\Vect{b}$ es una suma de múltiplos de esos vectores. Este dibujo muestra $2\Vect{a}+3\Vect{b}$.



```python
dibuja(2*a, 3*b, encadenados=True)
```

Cambie los dos coeficientes hasta que la suma sea el vector $(6,\,5,)$. Use la leyenda para saber dónde termina la flecha negra.



Los coeficientes pueden ser negativos o fracciones. Para escribir la fracción $\frac{1}{2}$ use `fracc(1,2)`. Dibuje $\frac{1}{2}\Vect{a}-2\Vect{b}$.



## Un sistema de ecuaciones visto por columnas



Esta forma de ver un sistema se explica en clase. Considere el sistema

\begin{equation*}
\begin{cases}
x+3y=6 \\
2x-y=5
\end{cases}
\end{equation*}

Los coeficientes de $x$ forman el vector $(1,\,2,)$. Los coeficientes de $y$ forman el vector $(3,\,-1,)$. Los números del lado derecho forman el vector $(6,\,5,)$. Son los vectores $\Vect{a}$ y $\Vect{b}$ de este Notebook. El sistema pregunta qué combinación lineal de $\Vect{a}$ y $\Vect{b}$ es igual a $(6,\,5,)$:

\begin{equation*}
x\,\Vect{a}+y\,\Vect{b}=\begin{pmatrix}6\\5\end{pmatrix}.
\end{equation*}

Usted ha encontrado esos coeficientes en el apartado anterior. Compruebe con NAcAL, sin dibujar, que la combinación lineal da el vector $(6,\,5,)$. Compruébelo también como producto de una matriz por un vector: la matriz es `Matrix([a, b])`.

## Qué debe saber al terminar



-   Un vector de $\R[2]$ se dibuja como una flecha desde el origen.
-   Los múltiplos de un vector están sobre una misma recta que pasa por el origen.
-   Para sumar vectores se pone cada flecha a continuación de la anterior. El orden no cambia el resultado.
-   Resolver un sistema de dos ecuaciones con dos incógnitas es buscar la combinación lineal de sus dos columnas que da el vector del lado derecho.
