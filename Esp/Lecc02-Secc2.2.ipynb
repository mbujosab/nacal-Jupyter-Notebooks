## Producto de una matriz por un vector



Lea la [Sección 2.2.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.2.2) hasta el primer ejercicio. El producto $\Mat{A}\Vect{b}$ es la combinación lineal de las columnas de $\Mat{A}$ cuyos coeficientes son las componentes de $\Vect{b}$. La matriz y el vector siguientes son los del ejemplo del libro.



```python
A = Matrix([ Vector([1,1,1]), Vector([1,0,1]), Vector([-2,2,1]) ])
b = Vector([1, 2, -3])
A
```

La primera celda escribe la combinación lineal de las columnas. La segunda escribe el producto.



```python
(A|1)*(b|1) + (A|2)*(b|2) + (A|3)*(b|3)
```

```python
A * b
```

Las dos celdas dan el mismo vector. La primera es la definición del producto.

Escriba la combinación lineal $2\Vect{x}+2\Vect{y}+2\Vect{z}$ del apartado anterior como producto de una matriz por un vector. La matriz cuyas columnas son esos tres vectores es `Matrix([x, y, z])`.



Escriba ahora como producto de una matriz por un vector la combinación lineal $2\,\VectC{A}{1}-\VectC{A}{3}$. Calcúlela antes con las columnas y compare los dos resultados.



La matriz $\Mat{A}$ tiene tres columnas. El vector debe tener tres componentes, una por cada columna. Ejecute la celda siguiente y lea la última línea del error. Es el mensaje del Notebook anterior. Aquí dice que la matriz y el vector no se pueden multiplicar.



```python
A * Vector([1, 2])
```

La matriz siguiente tiene dos filas y tres columnas. El vector tiene tres componentes, una por cada columna.



```python
C = Matrix([ [1,0,2], [3,1,1] ])
C * Vector([1, 2, 3])
```

El producto tiene dos componentes. Es una combinación lineal de las columnas, que son vectores de $\R[2]$. El producto $\Mat{A}\Vect{b}$ tiene tantas componentes como filas tiene $\Mat{A}$.

### Cada componente es un producto punto



Lea el resto de la [Sección 2.2.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.2.2). La componente $i$-ésima del vector $\Mat{A}\Vect{b}$ es el producto punto de la fila $i$-ésima de $\Mat{A}$ por $\Vect{b}$.



```python
2|(A*b)
```

```python
(2|A) * b
```

Las dos celdas dan el mismo número. La primera selecciona la segunda componente del producto. La segunda multiplica la segunda fila de $\Mat{A}$ por $\Vect{b}$. Compruebe lo mismo con la primera componente.



Python calcula el producto antes que la selección. Por eso `2|A*b` es lo mismo que `2|(A*b)`. El número es el mismo que el de `(2|A)*b`: es la igualdad que acaba de comprobar. La salida no dice en qué orden ha operado Python. En las dos celdas siguientes sí se ve.



```python
A*b|2
```

```python
A*(b|2)
```

La primera celda calcula $\Mat{A}\Vect{b}$ y selecciona su segunda componente. La segunda multiplica la matriz por el número $b_2$.

Si la matriz tiene una sola fila, el producto es un vector de $\R[1]$. No es un número.



```python
F = Matrix([ [1, 2, 3] ])
F * b
```

```python
(1|F) * b
```

La primera celda da un vector con una componente. La segunda da un número.



## Propiedades del producto de una matriz por un vector



Lea la [Sección 2.2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.2.3). Las celdas siguientes usan además esta matriz y este vector.



```python
B = Matrix([ [1,2,0], [0,1,3], [2,2,1] ])
c = Vector([2, 1, -1])
B
```

### Lineal en el vector



Distributivo respecto a la suma: $\Mat{A}(\Vect{b}+\Vect{c})=\Mat{A}\Vect{b}+\Mat{A}\Vect{c}$.



```python
A * (b + c) == A*b + A*c
```

Asociativo respecto al producto por escalares: $\Mat{A}(\lambda\Vect{b})=\lambda(\Mat{A}\Vect{b})$. Escríbalo usted.



Con el escalar a la derecha, esta propiedad es $\Mat{A}(\Vect{b}\lambda)=(\Mat{A}\Vect{b})\lambda$. Solo se desplazan los paréntesis.



```python
A * (b*3) == (A*b) * 3
```

### Una matriz por el producto de otra matriz por un vector



El corolario de la [Sección 2.2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.2.3) dice que $\Mat{A}(\Mat{B}\Vect{c})$ es el producto por $\Vect{c}$ de una matriz nueva. Las columnas de esa matriz son los productos de $\Mat{A}$ por cada columna de $\Mat{B}$.



```python
A * (B * c)
```

```python
Matrix([ A*(B|1), A*(B|2), A*(B|3) ]) * c
```

Las dos celdas dan el mismo vector. Muestre la matriz de la segunda celda sin multiplicarla por $\Vect{c}$. Compare cada columna con `A*(B|1)`, `A*(B|2)` y `A*(B|3)`. En la Lección 3 esta matriz será el producto $\Mat{A}\Mat{B}$.



### Lineal en la matriz



Distributivo respecto a la suma: $(\Mat{A}+\Mat{B})\Vect{c}=\Mat{A}\Vect{c}+\Mat{B}\Vect{c}$.



```python
(A + B) * c == A*c + B*c
```

Asociativo respecto al producto por escalares: $(\lambda\Mat{A})\Vect{b}=\lambda(\Mat{A}\Vect{b})$. Escríbalo usted.



La celda siguiente es la proposición del ejercicio del libro: $\Mat{A}(\lambda\Vect{b})=(\lambda\Mat{A})\Vect{b}$. Con la igualdad anterior da $(\lambda\Mat{A})\Vect{b}=\Mat{A}(\lambda\Vect{b})=\lambda(\Mat{A}\Vect{b})$.



```python
A * (3*b) == (3*A) * b
```

### El escalar puede ir en cualquier lugar



```python
3 * A * b
```

```python
A * 3 * b
```

```python
A * b * 3
```

Las tres celdas dan el mismo vector. No hacen falta paréntesis. Lo que no cambia es que el vector está a la derecha de la matriz.



## Qué debe saber al terminar



-   Una combinación lineal de vectores es una suma de múltiplos de esos vectores.
-   El producto $\Mat{A}\Vect{b}$ es una combinación lineal de las columnas de $\Mat{A}$. Los coeficientes son las componentes de $\Vect{b}$.
-   El número de componentes de $\Vect{b}$ es el número de columnas de $\Mat{A}$. El producto tiene tantas componentes como filas tiene $\Mat{A}$.
-   $\Mat{I}\Vect{a}=\Vect{a}$. Una matriz por la columna $j$-ésima de la identidad es la columna $j$-ésima de la matriz.
-   La componente $i$-ésima de $\Mat{A}\Vect{b}$ es el producto punto de la fila $i$-ésima de $\Mat{A}$ por $\Vect{b}$.
-   El producto de una matriz por un vector es lineal en el vector y lineal en la matriz.
-   $\Mat{A}(\Mat{B}\Vect{c})$ es el producto por $\Vect{c}$ de la matriz cuyas columnas son los productos de $\Mat{A}$ por cada columna de $\Mat{B}$.
-   El escalar puede ir en cualquier lugar del producto. El vector va siempre a la derecha de la matriz.
-   Python calcula el producto antes que la selección: `A*b|2` es la segunda componente de `A*b`.
