## Tres expresiones que se escriben sin paréntesis



Lea el comienzo de la [Sección 3.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.2). Las tres expresiones del recuadro se pueden leer de dos maneras: como la selección de una componente o como un producto punto. Las dos lecturas dan el mismo número. Las celdas de este Notebook usan estas matrices y estos vectores.



```python
A = Matrix([ [1,2,0], [3,-1,1] ])
B = Matrix([ [2,1], [0,-1], [1,3] ])
a = Vector([2, 1, -1])
b = Vector([1, -2, 2])
c = Vector([3, -1])
```

Primera: la componente $i$-ésima de $\Mat{A}\Vect{b}$ es el producto punto de la fila $i$-ésima de $\Mat{A}$ por $\Vect{b}$. La comprobó en el Notebook de la Sección 2.2.



```python
2|(A*b)
```

```python
(2|A) * b
```

Segunda: la componente $j$-ésima de $\Vect{a}\Mat{B}$ es el producto punto de $\Vect{a}$ por la columna $j$-ésima de $\Mat{B}$. Escriba las dos lecturas con $j=2$.



Tercera: la componente de la fila $i$ y la columna $j$ de $\Mat{AB}$ es el producto punto de la fila $i$-ésima de $\Mat{A}$ por la columna $j$-ésima de $\Mat{B}$. Es la Proposición 3.1.8 del Notebook anterior. Escriba las dos lecturas con $i=1$ y $j=2$.



En papel no hacen falta los paréntesis. El punto se escribe cuando, al poner los paréntesis, quedan juntos dos vectores. En Python sí hacen falta: Python calcula el producto antes que la selección, y `2|A*b` es lo mismo que `2|(A*b)`. Para la otra lectura hay que escribir `(2|A)*b`.



## Un vector por una matriz por un vector



Proposición 3.2.1: $(\Vect{a}\Mat{B})\cdot\Vect{c}=\Vect{a}\cdot(\Mat{B}\Vect{c})$.



```python
(a*B) * c
```

```python
a * (B*c)
```

Las dos celdas dan el mismo número. Por eso el libro escribe $\Vect{a}\Mat{B}\Vect{c}$ sin paréntesis. En esa expresión no quedan juntos dos vectores, y por eso no lleva punto.



```python
a * B * c
```

## Transpuesta de un producto



Lea el último ejercicio de la [Sección 3.2](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.2). Proposición 3.2.2: $(\Mat{AB})^\T=(\MatT{B})(\MatT{A})$. En NAcAL la transpuesta de `A` es `~A`.



```python
~(A*B)
```

```python
(~B) * (~A)
```

Las dos celdas dan la misma matriz. Al transponer cambia el orden de los factores. Pruebe con el otro orden: $(\MatT{A})(\MatT{B})$.



Con estas matrices el producto existe, pero es una matriz de orden 3. No puede ser la transpuesta de $\Mat{AB}$, que es de orden 2.

El libro avisa de que la expresión $\Mat{A}\MatT{B}$ es ambigua. Escriba siempre los paréntesis: `~(A*B)` es la transpuesta del producto y `A*(~B)` es $\Mat{A}$ por la transpuesta de $\Mat{B}$.



### Dos matrices simétricas



El libro deduce de la Proposición 3.2.2 que $(\MatT{A})\Mat{A}$ y $\Mat{A}(\MatT{A})$ son simétricas para cualquier matriz $\Mat{A}$.



```python
(~A) * A
```

```python
((~A) * A).es_simetrica()
```

Calcule $\Mat{A}(\MatT{A})$ y pregunte a NAcAL si es simétrica.



Las dos son simétricas, pero no son del mismo orden. ¿De qué orden es cada una si $\Mat{A}$ es de orden $m$ por $n$?

## Qué debe saber al terminar



-   Las expresiones $\elemL{\Mat{A}\Vect{b}}{i}$, $\elemR{\Vect{a}\Mat{B}}{j}$ y $\elemR{\elemL{\Mat{AB}}{i}}{j}$ se escriben sin paréntesis, porque sus dos lecturas dan el mismo número.
-   $(\Vect{a}\Mat{B})\cdot\Vect{c}=\Vect{a}\cdot(\Mat{B}\Vect{c})$. Por eso se escribe $\Vect{a}\Mat{B}\Vect{c}$, sin paréntesis y sin punto.
-   $(\Mat{AB})^\T=(\MatT{B})(\MatT{A})$: al transponer un producto cambia el orden de los factores.
-   $(\MatT{A})\Mat{A}$ y $\Mat{A}(\MatT{A})$ son simétricas para cualquier matriz $\Mat{A}$.
-   En Python hay que escribir los paréntesis que en papel se omiten.
