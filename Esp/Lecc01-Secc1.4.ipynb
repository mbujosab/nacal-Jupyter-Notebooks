## Matrices de $\R[m\times n]$



### Definiendo matrices



La notación indicada en el [libro](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.1.4) establece que al encerrar entre corchetes una lista de vectores (con el mismo número de componentes) se denota la matriz cuyas columnas son dichos vectores. [NAcAL](https://mbujosab.github.io/nacallib/) implementa este convenio utilizando una lista o tupla de vectores (con el mismo número de componentes) como argumento de la clase `Matrix()`.

Así, al encerrar los vectores $(1, 2,)$, $(1, 0,)$, y $(9, 2,)$ entre corchetes se denota la matriz de $\R[2\times 3]$:

$$\begin{bmatrix}1&1&9\\2&0&2\end{bmatrix}$$



Con [NAcAL](https://mbujosab.github.io/nacallib/) logramos lo mismo colocando dicha lista de vectores como argumento de la clase `Matrix( )`



```python
a = Vector( [1,2] )
b = Vector( [1,0] )
c = Vector( [9,2] )

D = Matrix( [a,b,c] )
```

```python
print(D)              # representación de la matriz en consola
display(D)            # representación de la matriz en Jupyter Notebook
```

La representación con `repr(D)` indica explícitamente que `D` es una matriz formada por una lista de vectores, cada uno formado por una lista de números:



```python
repr(D)              # representación más explícita
```

#### Una forma de instanciar distinta de la del libro, pero más cómoda al trabajar con el ordenador



Podemos instanciar una matriz mediante una lista (o tupla) de listas o tuplas. En este caso, describimos la matriz fila a fila:



```python
A = Matrix( ( [1,1], [2,4], [3,9] ) )
A
```

Esta forma alternativa de definir matrices (*fila a fila* en lugar de *columna a columna*) resulta más cómoda al usar [NAcAL](https://mbujosab.github.io/nacallib/) (aunque no es la descrita en el [libro](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.1.4)).



### Algunos atributos y procedimientos de `Matrix()`



#### Atributo: `.lista`



[NAcAL](https://mbujosab.github.io/nacallib/) almacena la lista ordenada de vectores (*de columnas*) de cada `Matrix()` en su atributo `.lista`.



```python
print( A.lista )
```

Fíjese que, aunque definimos la matriz `A` *fila a fila*, el atributo `.lista` almacena la correspondiente lista de columnas (respetando de este modo la definición de matriz dada en el curso).



#### El orden de la matriz (atributos: `.m` y `.n`)



El atributo `.m` nos indica el número de filas de la matriz y el atributo `.n` su número de columnas. Así, el orden de $\Mat{A}$ es



```python
print( A.m )
print( A.n )
```

#### Método: `.sis()`



Podemos representar una matriz en la forma genérica de un `Sistema()` con `.sis()`.



```python
A.sis()
```

```python
print(A.sis())
```

Fíjese que `repr(A.sis())` indica que `A.sis()` es un sistema formado por una lista de vectores.



```python
repr(A.sis())
```

Y que `repr(A)` muestra que `A` es una matriz (es decir, un sistema) formada por idéntica lista de vectores.



```python
repr(A)
```

Así, aunque `A.sis()` y `A` son objetos de tipo distinto



```python
B = A.sis()
print(type(A))
print(type(B))
```

y con diferente representación en `NAcAL`,



```python
# Creamos un sistema con ambos objetos y cadenas de texto para ver A y B juntos
Sistema( ['La matriz: ', A, 'El sistema: ',B] )
```

en realidad, son el mismo objeto matemático (listas ordenadas de vectores). Por eso, si preguntamos a Python si `A` y `B` son iguales, nos responde que sí.



```python
B==A
```

## La transposición



(*[Sección 1.4.5](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.1.4.5) del libro*)

La transpuesta de $\Mat{D}$ es la matriz cuyas columnas son las filas de $\Mat{D}$. La matriz $\Mat{D}$ tiene dos filas. Con ellas formamos una matriz:



```python
Matrix( [ 1|D, 2|D ] )
```

Usando las herramientas descritas anteriormente, disponemos de otra forma de transponer $\Mat{A}$: 
introducir como argumento de `Matrix()` la lista de las listas correspondientes a sus columnas (que es la forma de describir en [NAcAL](https://mbujosab.github.io/nacallib/) una matriz fila a fila).

Así, si la matriz es: 
$$\Mat{D}=\begin{bmatrix}1&1&9\\2&0&2\end{bmatrix}$$ entonces su transpuesta es:



```python
Matrix( [ (D|1).lista, (D|2).lista, (D|3).lista ] )
```

Esto se puede automatizar con un bucle que recorra las columnas de la matriz (usando [list comprehension](https://ellibrodepython.com/list-comprehension-python)):



```python
B = Matrix ( [ (D|j).lista for j in range(1,D.n+1) ] )
B
```

Si estudia el código de [NAcAL](https://mbujosab.github.io/nacallib/), verá que la transposición está programada tal cual.



He buscado un símbolo en Python para indicar la operación de la *transposición* al estilo del símbolo $^\T$ usado en el curso. 

Desgraciadamente, Python tiene una fuerte limitación en cuanto a la disponibilidad de símbolos para ser usados como ["métodos mágicos"](https://realpython.com/python-magic-methods/), así que no he encontrado mejor solución que emplear el símbolo (`~`) para la transposición (en lugar del símbolo $^\T$). 

Desafortunadamente, dicho símbolo ha de situarse *delante* de la matriz que se quiere transponer. 
Por lo tanto, en [NAcAL](https://mbujosab.github.io/nacallib/), la transpuesta $\MatT{D}$ se escribe así: `~D`



```python
~D
```

Verifiquemos que al transponer cambian de lado los índices de los elementos de la matriz:



```python
1|(D)|3
```

```python
3|(~D)|1
```

```python
3|(~D) == D|3 # la fila 3 de D transpuesta es igual que la columna 3 de D
```

```python
(~D)|2 == 2|D # la columna 2 de D transpuesta es igual que la fila 2 de D
```

Si queremos obtener una matriz cuya única *columna* es la segunda fila de $\Mat{D}$, basta instanciar una `Matrix` usando una lista que contenga dicha fila:



```python
Matrix( [2|D] )
```

## Definiendo una matriz a partir de las filas y columnas de otra



(*Esta parte corresponde al [apéndice](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph2.3.Alph2) de la Lección 3 del libro. Allí se extiende a las matrices la selección con una lista de índices del [apéndice](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.alph1.1.Alph1) de la Lección 1*)

Cuando seleccionamos varias filas o varias columnas con una tupla obtenemos una submatriz. 
Veámoslo



```python
B = Matrix( ( [1,1,1], [2,4,8], [3,9,0] ) )
B
```

```python
B|(1,3,)
```

```python
(2,3,)|B
```

```python
(2,3,)|B|(1,3,)
```

```python
B|(1,1,1,1,1,)
```

Tenga en cuenta que en Python `(2)` es el número `2` entre paréntesis, pero `(2,)` es una tupla con un único elemento (el `2`). 

La diferencia es la coma "`,`" detrás del número. 

En el [libro](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf) se ha adoptado el mismo convenio de notación para poder distinguir entre números y vectores de $\R[1]$.



```python
type( (2) )
```

```python
type( (2,) )
```

`NAcAL` tiene un comportamiento muy distinto si el índice que selecciona una columna (o una fila) está dentro de una tupla o no lo está. Aquí el índice de la derecha es un número, por lo que seleccionamos la tercera columna, que es un vector:



```python
(3,2,1,)|B|3
```

Aquí el índice de la derecha está dentro de una tupla, por lo que creamos una *matriz* columna:



```python
(3,2,1,)|B|(3,)
```

```python
(2,)|D
```

## Qué debe saber al terminar



-   Una matriz es una lista ordenada de vectores con el mismo número de componentes. Esos vectores son sus columnas.
-   `A|j` es la columna $j$, `i|A` es la fila $i$ y `i|A|j` es la componente de la fila $i$ y la columna $j$.
-   Dos matrices son iguales si tienen las mismas columnas en el mismo orden.
-   Las columnas de la transpuesta son las filas de la matriz. Al transponer, los índices cambian de lado.
-   Una misma matriz puede ser de varios tipos a la vez.
