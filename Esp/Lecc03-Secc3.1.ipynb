## Producto de matrices



Lea la definición al comienzo de la [Sección 3.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.1) y el ejemplo que la sigue. La columna $j$-ésima de $\Mat{AB}$ es el producto de $\Mat{A}$ por la columna $j$-ésima de $\Mat{B}$. Las matrices siguientes son las del ejemplo del libro.



```python
x = Vector([1, 1, -2])
y = Vector([1, -2, 1])
z = Vector([-2, 1, 1])
A = Matrix([x, y, z])
B = Matrix([z, x])
A
```

```python
B
```

La primera celda escribe el producto de $\Mat{A}$ por la segunda columna de $\Mat{B}$. La segunda celda escribe la segunda columna del producto $\Mat{AB}$.



```python
A * (B|2)
```

```python
(A*B)|2
```

Las dos celdas dan el mismo vector. Compruebe lo mismo con la primera columna.



Escriba ahora el producto completo y compare sus dos columnas con los vectores anteriores.



```python
A * B
```

Cada columna de $\Mat{AB}$ es una combinación lineal de las columnas de $\Mat{A}$. Los coeficientes son las componentes de la columna correspondiente de $\Mat{B}$.

En el producto del libro se repiten varios números. Por eso un error puede dar el resultado correcto. Las dos matrices siguientes dan un producto con todas sus componentes distintas.



```python
R = Matrix([ [1,2,0], [3,-1,4], [0,1,3] ])
S = Matrix([ [2,1], [0,-1], [1,3] ])
R * S
```

Escriba la segunda columna de $\Mat{RS}$ como combinación lineal de las tres columnas de $\Mat{R}$.



### Cuándo existe el producto



El producto solo está definido si la primera matriz tiene tantas columnas como filas tiene la segunda. $\Mat{A}$ tiene tres columnas y $\Mat{B}$ tiene tres filas. Por eso existe $\Mat{AB}$. Decida si existe $\Mat{BA}$ antes de ejecutar la celda siguiente.



```python
B * A
```

NAcAL da un error, porque $\Mat{B}$ tiene dos columnas y $\Mat{A}$ tiene tres filas.



## Propiedades del producto de matrices



Lea los dos ejercicios de la [Sección 3.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.1) que enuncian las proposiciones 3.1.1 a 3.1.7. Las celdas de este apartado usan además estas matrices y este vector. $\Mat{C}$ es del mismo orden que $\Mat{A}$, y $\Mat{D}$ es del mismo orden que $\Mat{B}$.



```python
C = Matrix([ [1,0,2], [0,1,-1], [3,1,0] ])
D = Matrix([ [0,1], [2,-1], [1,3] ])
F = Matrix([ [2,1], [-1,3] ])
v = Vector([1, 2])
```

### El producto es asociativo



Proposición 3.1.1: $\Mat{A}(\Mat{B}\Vect{v})=(\Mat{AB})\Vect{v}$.



```python
A * (B*v)
```

```python
(A*B) * v
```

Las dos celdas dan el mismo vector. Proposición 3.1.2: $\Mat{A}(\Mat{BF})=(\Mat{AB})\Mat{F}$. Compruébela.



Por eso el libro escribe $\Mat{ABF}$ sin paréntesis. Python calcula `A*B*F` de izquierda a derecha, y da la misma matriz.



```python
A * B * F
```

### El producto es lineal



Proposición 3.1.3: $(\Mat{A}+\Mat{C})\Mat{B}=\Mat{AB}+\Mat{CB}$.



```python
(A + C) * B == A*B + C*B
```

Proposición 3.1.4: $\Mat{A}(\Mat{B}+\Mat{D})=\Mat{AB}+\Mat{AD}$. Escríbala usted.



Proposición 3.1.5: $\Mat{A}(\lambda\Mat{B})=\lambda(\Mat{AB})=(\lambda\Mat{A})\Mat{B}$. La celda comprueba la primera igualdad con $\lambda=5$. Escriba usted la segunda.



```python
A * (5*B) == 5 * (A*B)
```

El libro resume estas tres proposiciones así: el producto de matrices es un operador lineal, y también es lineal por la izquierda.



### La matriz identidad



Proposiciones 3.1.6 y 3.1.7: $\Mat{IB}=\Mat{B}$ y $\Mat{BI}=\Mat{B}$. $\Mat{B}$ tiene tres filas y dos columnas. Por eso la identidad de la primera igualdad es de orden 3.



```python
I(3) * B == B
```

¿De qué orden es la identidad de la segunda igualdad? Compruébelo.



## El producto de matrices no es conmutativo



Lea el párrafo «Muy muy importante» de la [Sección 3.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.3.1). En general $\Mat{AB}$ es distinto de $\Mat{BA}$. Las dos matrices siguientes son las del libro.



```python
G = Matrix([ [1,1], [0,1] ])
H = Matrix([ [1,0], [0,3] ])
G * H
```

```python
H * G
```

Las dos matrices son distintas en una componente. Con matrices que no son cuadradas puede existir uno de los dos productos y no el otro. Es lo que ocurrió con $\Mat{BA}$ al principio de este Notebook.

El libro da otras dos propiedades de los números que las matrices no tienen. Entre los números, si $\lambda^2=\lambda$ entonces $\lambda$ es 0 o 1. La matriz siguiente es igual a su cuadrado y no es la matriz nula ni la identidad.



```python
P = Matrix([ [1,0], [0,0] ])
P * P == P
```

Entre los números, si $\lambda^2=0$ entonces $\lambda$ es 0. Calcule el cuadrado de la matriz siguiente.



```python
N = Matrix([ [1,1], [-1,-1] ])
N
```

El cuadrado es la matriz nula, y $\Mat{N}$ no lo es.



## Otras dos formas de calcular el producto



Lea la [Sección 3.1.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.3.1.1). Sus dos proposiciones se deducen de la definición de producto por columnas.



### Componente a componente



Proposición 3.1.8: la componente de la fila $i$ y la columna $j$ de $\Mat{AB}$ es el producto punto de la fila $i$-ésima de $\Mat{A}$ por la columna $j$-ésima de $\Mat{B}$. Las celdas usan las matrices $\Mat{R}$ y $\Mat{S}$.



```python
2|(R*S)|1
```

```python
(2|R) * (S|1)
```

Las dos celdas dan el mismo número. Compruebe lo mismo con la componente de la fila 3 y la columna 2.



Esta es la forma habitual de calcular el producto a mano: filas por columnas.



### Fila a fila



Proposición 3.1.9: la fila $i$-ésima de $\Mat{AB}$ es el producto de la fila $i$-ésima de $\Mat{A}$ por la matriz $\Mat{B}$.



```python
2|(R*S)
```

```python
(2|R) * S
```

Las dos celdas dan el mismo vector. Compruebe lo mismo con la tercera fila.



Por tanto, cada fila de $\Mat{AB}$ es una combinación lineal de las filas de $\Mat{B}$. Los coeficientes son las componentes de la fila correspondiente de $\Mat{A}$. Escriba la segunda fila de $\Mat{RS}$ como combinación lineal de las tres filas de $\Mat{S}$.



El libro pide verificar las dos proposiciones con el producto de su ejemplo. Es el producto $\Mat{AB}$ del comienzo de este Notebook. Calcúlelo a mano de las tres formas: columna a columna, componente a componente y fila a fila. Compare cada resultado con la salida de la celda `A * B`.



## Qué debe saber al terminar



-   La columna $j$-ésima de $\Mat{AB}$ es $\Mat{A}$ por la columna $j$-ésima de $\Mat{B}$. Las columnas de $\Mat{AB}$ son combinaciones lineales de las columnas de $\Mat{A}$.
-   El producto solo existe si $\Mat{A}$ tiene tantas columnas como filas tiene $\Mat{B}$.
-   El producto es asociativo. Es un operador lineal y también es lineal por la izquierda.
-   $\Mat{IA}=\Mat{A}$ y $\Mat{AI}=\Mat{A}$, con la identidad del orden adecuado a cada lado.
-   En general $\Mat{AB}$ es distinto de $\Mat{BA}$.
-   La componente de la fila $i$ y la columna $j$ de $\Mat{AB}$ es el producto punto de la fila $i$-ésima de $\Mat{A}$ por la columna $j$-ésima de $\Mat{B}$.
-   La fila $i$-ésima de $\Mat{AB}$ es la fila $i$-ésima de $\Mat{A}$ por $\Mat{B}$. Las filas de $\Mat{AB}$ son combinaciones lineales de las filas de $\Mat{B}$.
