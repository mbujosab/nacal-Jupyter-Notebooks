## Eliminación de izquierda a derecha



Lea la [Sección 4.2.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.2.1). Pre-escalonar una matriz es conseguir, con transformaciones elementales, que las componentes a la derecha de cada pivote sean nulas. Las columnas se recorren de izquierda a derecha. La matriz siguiente es la del ejemplo del libro.



```python
A = Matrix([ [0,1,1], [2,-2,2], [2,-8,-4] ])
A
```

El pivote de la primera columna es el 2 de la segunda fila. A su derecha hay un $-2$ y un 2. Para anularlos se suma la primera columna a la segunda y se resta la primera columna de la tercera. Las transformaciones se aplican a una copia de $\Mat{A}$, llamada `M`.



```python
M = Matrix(A)
M & T( (1,1,2) ) & T( (-1,1,3) )
```

Ahora el pivote de la segunda columna es el 1 de la primera fila. A su derecha hay un 1. Se resta la segunda columna de la tercera.



```python
M & T( (-1,2,3) )
```

La matriz es pre-escalonada. Su tercera columna es nula. Si ejecuta dos veces una de estas celdas, la transformación se aplica dos veces. En ese caso ejecute de nuevo la celda que define `M` y repita los pasos.

`Elim` hace la eliminación de izquierda a derecha. Con un 1 como segundo argumento muestra los pasos.



```python
Elim(A, 1)
```

Los pasos son los mismos. `Elim` no cambia la matriz $\Mat{A}$: devuelve una matriz nueva.

### Fracciones



En la matriz siguiente los pivotes no son unos.



```python
C = Matrix([ [2,1,3], [3,5,-1], [4,-2,7] ])
Elim(C, 1)
```

El pivote de la primera columna es un 2 y a su derecha hay un 1. Para anularlo con una sola transformación habría que restar de la segunda columna media primera columna. NAcAL evita la fracción: primero multiplica la segunda columna por 2 y después le resta la primera. Multiplicar una columna por 2 es una transformación de Tipo II. El libro lo explica después del Teorema 4.2.1.

La demostración del Teorema 4.2.1 anula las componentes con transformaciones de Tipo I. NAcAL también tiene esa variante. El último apartado de este Notebook explica las variantes.



```python
C.elim(1, rep=True)
```

Las dos matrices obtenidas son formas pre-escalonadas de $\Mat{C}$ y no son iguales. Una matriz tiene muchas formas pre-escalonadas. Compare los pivotes de las dos: ¿están en las mismas filas?

Pre-escalone $\Mat{C}$ en papel de la manera que prefiera. Después compruebe sus pasos con la copia `N`. En NAcAL una fracción se escribe con `fracc`: `fracc(1,2)` es un medio.



```python
N = Matrix(C)
```

## Eliminación Gauss-Jordan



Lea la [Sección 4.2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.4.2.3). Una matriz escalonada reducida es una matriz escalonada en la que cada pivote es un 1 y las demás componentes de la fila de cada pivote son nulas.

La celda siguiente guarda en `L` una forma escalonada de $\Mat{B}$.



```python
L = ElimG(B)
L
```

La demostración del Corolario 4.2.3 la reduce en este orden. Se empieza por el pivote de la primera columna. Con cada pivote se anulan las componentes que tiene a su izquierda. Al final se divide cada columna por su pivote. El primer pivote no tiene nada a su izquierda. El segundo pivote es el 1 de la segunda fila y a su izquierda hay un 2.



```python
L & T( (-2,2,1) )
```

El tercer pivote es el 5 de la tercera fila. Anule las dos componentes que tiene a su izquierda. Después divida la tercera columna por 5. Necesitará `fracc`.



`ElimGJ` hace la eliminación Gauss-Jordan.



```python
ElimGJ(B, 1)
```

La matriz final es la misma que la suya, pero los pasos no son los mismos. El libro avisa de que el orden de las transformaciones no es fijo. NAcAL evita las fracciones y deja la reordenación de las columnas para el final. El último apartado de este Notebook trata de eso.

La celda siguiente aplica la eliminación Gauss-Jordan a la matriz $\Mat{A}$ del principio.



```python
ElimGJ(A, 1)
```

### Las variantes de la eliminación



`Elim`, `ElimG` y `ElimGJ` usan un mismo procedimiento, llamado `elim`, que tiene variantes. Cada variante tiene un número. Para combinar varias se suman sus números.


| Número|Qué hace|
|---|---|
| 0|Pre-escalona de izquierda a derecha y evita las fracciones, como <code>Elim</code>|
| +1|Usa solo transformaciones de Tipo I, aunque aparezcan fracciones|
| +2|Anula también las componentes que están a la izquierda de cada pivote|
| +10|Divide cada columna por su pivote|
| +20|Reordena las columnas para que la matriz sea escalonada|
| +100|Recorre las columnas de derecha a izquierda|

Con `rep=True` se muestran los pasos. La variante 2 anula las componentes a los dos lados de cada pivote.



```python
B.elim(2, rep=True)
```

La suma $2+10+20=32$ es la eliminación Gauss-Jordan.



```python
B.elim(32, rep=True)
```

Escriba la variante que usa solo transformaciones de Tipo I y además reordena las columnas. Aplíquela a la matriz `A2` del apartado de problemas.



La variante 100 recorre las columnas de derecha a izquierda. En cada columna usa como pivote la última componente no nula y anula las componentes que ese pivote tiene a su izquierda.



```python
B.elim(100, rep=True)
```

El resultado no es una matriz pre-escalonada con la definición del libro, porque aquí el pivote es otro. Cuente sus columnas no nulas.

Pre-escalone $\Mat{B}$ con las variantes 0, 1, 2 y 20 y cuente en cada caso las columnas no nulas. Siempre son tres. La Lección 5 demuestra que todas las formas pre-escalonadas de una matriz tienen el mismo número de pivotes. A ese número lo llama rango de la matriz. Con $\Mat{B}$ las variantes 0 y 1 dan la misma matriz. ¿Por qué?



### Encadenar etapas



`elim` devuelve una matriz, y a esa matriz se le puede aplicar otra variante. Así se decide el orden de las etapas. La celda siguiente sigue el orden de la demostración del Corolario 4.2.3. Pre-escalona. Reordena las columnas. Anula las componentes a la izquierda de los pivotes. Divide cada columna por su pivote.



```python
B.elim(0).elim(20).elim(2).elim(10, rep=True)
```

Compare con los pasos de `ElimGJ(B, 1)`. El orden de los pasos es distinto y la matriz final es la misma.

## Qué debe saber al terminar



-   El pivote de una columna no nula es su primera componente no nula.
-   Una matriz es pre-escalonada si las componentes a la derecha de cada pivote son nulas. Es escalonada si además las columnas nulas están a la derecha y cada pivote está más abajo que el de la columna anterior. Es escalonada reducida si además cada pivote es un 1 y las demás componentes de su fila son nulas.
-   Toda matriz se puede pre-escalonar, escalonar y reducir con una secuencia de transformaciones elementales de sus columnas (Teorema 4.2.1 y Corolarios 4.2.2 y 4.2.3).
-   Para pre-escalonar a mano se recorren las columnas de izquierda a derecha. Con el pivote de cada columna se anulan las componentes que tiene a su derecha. Una columna nula se deja como está.
-   Una matriz tiene muchas formas pre-escalonadas. NAcAL evita las fracciones con transformaciones de Tipo II, y por eso sus pasos pueden no coincidir con los que usted da en papel.
-   `Elim`, `ElimG` y `ElimGJ` sirven para comprobar lo que antes se ha hecho en papel.
