## Las funciones de dibujo



NAcAL no dibuja. La celda siguiente define tres funciones que dibujan en $\R[3]$: `dibuja_combinaciones`, `dibuja_filas` y `dibuja_columnas`. Usan la librería [Plotly](https://plotly.com/python/). No hace falta entender su código. Ejecute la celda y pase al apartado siguiente.



```python
import numpy as np
import plotly.graph_objects as go

COLORES = ['royalblue', 'firebrick', 'seagreen']

def _numeros(v):
    if not isinstance(v, Vector) or v.n != 3:
        raise ValueError('Hace falta un vector de R^3.')
    return np.array([float(i|v) for i in (1, 2, 3)])

def _tabla(A):
    if not isinstance(A, Matrix) or A.m != 3 or A.n != 3:
        raise ValueError('Hace falta una matriz de orden 3 por 3.')
    return np.array([[float(i|A|j) for j in (1, 2, 3)] for i in (1, 2, 3)])

def _texto(p):
    return '(' + ', '.join('%g' % (round(float(c), 4) + 0.0) for c in p) + ',)'

def _flecha(fig, inicio, fin, color, nombre, grosor=8, punta=0.5):
    inicio = np.array(inicio, float)
    fin = np.array(fin, float)
    d = fin - inicio
    largo = np.linalg.norm(d)
    fig.add_trace(go.Scatter3d(x=[inicio[0], fin[0]], y=[inicio[1], fin[1]],
                               z=[inicio[2], fin[2]], mode='lines',
                               line=dict(color=color, width=grosor),
                               name=nombre, legendgroup=nombre))
    if largo < 1e-9:
        return
    d = d / largo
    u = np.cross(d, [0, 0, 1])
    if np.linalg.norm(u) < 1e-6:
        u = np.cross(d, [0, 1, 0])
    u = u / np.linalg.norm(u)
    w = np.cross(d, u)
    alto = min(punta, largo / 2)
    base = fin - alto * d
    k = 12
    aro = [base + 0.3 * alto * (np.cos(t) * u + np.sin(t) * w)
           for t in np.linspace(0, 2 * np.pi, k, endpoint=False)]
    p = np.array([fin, base] + aro)
    fig.add_trace(go.Mesh3d(x=p[:, 0], y=p[:, 1], z=p[:, 2],
                            i=[0] * k + [1] * k,
                            j=[2 + n for n in range(k)] * 2,
                            k=[2 + (n + 1) % k for n in range(k)] * 2,
                            color=color, flatshading=True, hoverinfo='skip',
                            legendgroup=nombre, showlegend=False))

def _trozo_de_plano(fig, centro, u, w, color, nombre):
    centro = np.array(centro, float)
    p = np.array([centro - u - w, centro + u - w, centro + u + w, centro - u + w])
    fig.add_trace(go.Mesh3d(x=p[:, 0], y=p[:, 1], z=p[:, 2],
                            i=[0, 0], j=[1, 2], k=[2, 3],
                            color=color, opacity=0.45, flatshading=True,
                            hoverinfo='skip', name=nombre, showlegend=True))
    borde = np.vstack([p, p[:1]])
    fig.add_trace(go.Scatter3d(x=borde[:, 0], y=borde[:, 1], z=borde[:, 2],
                               mode='lines', line=dict(color=color, width=2),
                               hoverinfo='skip', showlegend=False))

def _termina(fig, titulo, figura):
    fig.add_trace(go.Scatter3d(x=[0], y=[0], z=[0], mode='markers',
                               marker=dict(color='black', size=4),
                               name='origen'))
    fig.update_layout(title=titulo, height=650,
                      margin=dict(l=0, r=0, t=50, b=0),
                      scene=dict(aspectmode='data', xaxis_title='x',
                                 yaxis_title='y', zaxis_title='z',
                                 camera=dict(eye=dict(x=2.0, y=0.3, z=0.9))))
    if figura:
        return fig
    fig.show()

def dibuja_combinaciones(v, w, x=1, y=1, figura=False):
    """Dibuja dos vectores de R^3, el trozo de plano que llenan sus
    combinaciones lineales y la combinación lineal x*v + y*w."""
    va, vb = _numeros(v), _numeros(w)
    x, y = float(x), float(y)
    fig = go.Figure()
    alcance = max(1.5, abs(x) + 0.5, abs(y) + 0.5)
    _trozo_de_plano(fig, [0, 0, 0], alcance * va, alcance * vb, 'goldenrod',
                    'combinaciones lineales de v y w')
    punta = 0.06 * max(np.linalg.norm(va), np.linalg.norm(vb), 1)
    _flecha(fig, [0, 0, 0], va, COLORES[0], 'v = ' + _texto(va), punta=punta)
    _flecha(fig, [0, 0, 0], vb, COLORES[1], 'w = ' + _texto(vb), punta=punta)
    suma = x * va + y * vb
    _flecha(fig, x * va, suma, 'lightcoral', '%g w, desde la punta de %g v' % (y, x),
            grosor=4, punta=punta)
    _flecha(fig, [0, 0, 0], suma, 'black',
            '%g v %s %g w = ' % (x, '+' if y >= 0 else '-', abs(y)) + _texto(suma), grosor=5, punta=punta)
    return _termina(fig, 'Combinaciones lineales de dos vectores', figura)

def dibuja_filas(A, b, figura=False):
    """Visión por filas del sistema Ax = b: dibuja el plano de cada
    ecuación y, si la solución es única, el punto donde se cortan."""
    M, lado = _tabla(A), _numeros(b)
    fig = go.Figure()
    solucion = np.linalg.solve(M, lado) if abs(np.linalg.det(M)) > 1e-9 else None
    tamano = 3 if solucion is None else 3 + np.linalg.norm(solucion) / 2
    for i in range(3):
        n = M[i]
        if np.linalg.norm(n) < 1e-9:
            print('La ecuación', i + 1, 'no tiene incógnitas: no es un plano.')
            continue
        n1 = n / np.linalg.norm(n)
        centro = lado[i] * n / n.dot(n)
        if solucion is not None:
            centro = solucion
        u = np.cross(n1, [0, 0, 1])
        if np.linalg.norm(u) < 1e-6:
            u = np.cross(n1, [0, 1, 0])
        u = u / np.linalg.norm(u)
        w = np.cross(n1, u)
        _trozo_de_plano(fig, centro, tamano * u, tamano * w, COLORES[i],
                        'ecuación %d' % (i + 1))
    if solucion is None:
        titulo = 'Visión por filas: los tres planos no se cortan en un único punto'
    else:
        fig.add_trace(go.Scatter3d(x=[solucion[0]], y=[solucion[1]], z=[solucion[2]],
                                   mode='markers', marker=dict(color='black', size=7),
                                   name='solución ' + _texto(solucion)))
        titulo = 'Visión por filas: los tres planos se cortan en ' + _texto(solucion)
    return _termina(fig, titulo, figura)

def dibuja_columnas(A, b, figura=False):
    """Visión por columnas del sistema Ax = b: dibuja las columnas de A,
    el vector b y, si la solución es única, la combinación lineal de
    las columnas que da b."""
    M, lado = _tabla(A), _numeros(b)
    fig = go.Figure()
    punta = 0.06 * max(np.linalg.norm(lado), max(np.linalg.norm(M[:, j]) for j in range(3)), 1)
    _flecha(fig, [0, 0, 0], lado, 'black', 'b = ' + _texto(lado), grosor=12, punta=punta)
    for j in range(3):
        _flecha(fig, [0, 0, 0], M[:, j], COLORES[j],
                'columna %d = ' % (j + 1) + _texto(M[:, j]), punta=punta)
    if abs(np.linalg.det(M)) > 1e-9:
        solucion = np.linalg.solve(M, lado)
        inicio = np.zeros(3)
        for j in range(3):
            fin = inicio + solucion[j] * M[:, j]
            _flecha(fig, inicio, fin, COLORES[j],
                    '%g por la columna %d' % (round(solucion[j], 4) + 0.0, j + 1),
                    grosor=4, punta=punta)
            inicio = fin
        titulo = 'Visión por columnas: b es la combinación lineal con coeficientes ' + _texto(solucion)
    else:
        titulo = 'Visión por columnas: no hay una única combinación lineal que dé b'
    return _termina(fig, titulo, figura)
```

## Detalles de Python



-   Las tres funciones solo admiten vectores de $\R[3]$ y matrices de orden 3 por 3.
-   `dibuja_combinaciones(v, w, x, y)` dibuja la combinación lineal con coeficientes `x` e `y`. Sin ellos usa 1 y 1.
-   Las funciones calculan la solución con números decimales para poder dibujarla. Por eso el título puede mostrar un valor redondeado.
-   Con el argumento `figura=True` la función no dibuja y devuelve la figura de Plotly.



## Qué debe saber al terminar



-   Las combinaciones lineales de dos vectores de $\R[3]$ que no son múltiplos uno del otro llenan un plano que pasa por el origen.
-   En la visión por filas cada ecuación es un plano. La solución es el punto común a los tres planos.
-   En la visión por columnas la solución da los coeficientes de la combinación lineal de las columnas que es igual al lado derecho.
-   Si las tres columnas están en un mismo plano, solo tienen solución los sistemas cuyo lado derecho está en ese plano.
