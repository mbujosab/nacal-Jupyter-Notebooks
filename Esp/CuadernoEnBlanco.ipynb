## Recordatorio de órdenes



Un vector.



```python
a = Vector([1, 2, 3])
a
```

Una matriz, escrita fila a fila.



```python
A = Matrix([[1, 2, 3], [2, 4, 7], [1, 1, 1]])
A
```

El selector por la derecha elige una columna de la matriz.



```python
A|2
```

El selector por la izquierda elige una fila.



```python
1|A
```

Con los dos selectores se elige una componente.



```python
1|A|2
```

Una fracción.



```python
Vector([fracc(1, 3), 2])
```

Variables simbólicas.



```python
x, y = sympy.symbols('x y')
Vector([x, y, 1])
```

Eliminación hasta una forma pre-escalonada. Con el segundo argumento se muestran los pasos.



```python
Elim(A, 1)
```

Eliminación gaussiana, hasta una forma escalonada.



```python
ElimG(A, 1)
```

Eliminación Gauss-Jordan, hasta la forma escalonada reducida.



```python
ElimGJ(A, 1)
```

La matriz inversa.



```python
A.inversa()
```

El rango.



```python
A.rango()
```
