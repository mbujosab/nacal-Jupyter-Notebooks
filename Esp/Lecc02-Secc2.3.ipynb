## Producto de un vector por una matriz



Lea la definición al comienzo de la [Sección 2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.3). El producto $\Vect{b}\Mat{A}$ es la combinación lineal de las filas de $\Mat{A}$ cuyos coeficientes son las componentes de $\Vect{b}$. La matriz y el vector son los del Notebook anterior.



```python
A = Matrix([ Vector([1,1,1]), Vector([1,0,1]), Vector([-2,2,1]) ])
b = Vector([1, 2, -3])
A
```

La primera celda escribe la combinación lineal de las filas. La segunda escribe el producto.



```python
(b|1)*(1|A) + (b|2)*(2|A) + (b|3)*(3|A)
```

```python
b * A
```

Las dos celdas dan el mismo vector. Calcule ahora $\Mat{A}\Vect{b}$ y compárelo con $\Vect{b}\Mat{A}$.

## Un vector por una matriz es la transpuesta de la matriz por el vector



Las filas de $\Mat{A}$ son las columnas de su transpuesta. Por eso $\Vect{b}\Mat{A}=(\MatT{A})\Vect{b}$. El libro lo demuestra en la [Sección 2.3](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#section.2.3), después de la definición. En NAcAL la transpuesta de `A` es `~A`.



```python
b * A == (~A) * b
```

El libro enuncia a continuación la igualdad para el otro producto: $\Mat{A}\Vect{b}=\Vect{b}(\MatT{A})$. Sale de aplicar la igualdad anterior a la matriz $\MatT{A}$, porque la transpuesta de $\MatT{A}$ es $\Mat{A}$. Compruébela.



## Cada componente es un producto punto



La componente $j$-ésima del vector $\Vect{b}\Mat{A}$ es el producto punto de $\Vect{b}$ por la columna $j$-ésima de $\Mat{A}$.



```python
(b*A)|3
```

```python
b * (A|3)
```

Las dos celdas dan el mismo número. Compruebe lo mismo con la segunda componente.



## Propiedades del producto de un vector por una matriz



Lea la [Sección 2.3.1](https://mbujosab.github.io/CursoDeAlgebraLineal/libro.pdf#subsection.2.3.1). Su ejercicio enuncia seis propiedades. Tres están escritas aquí. Escriba usted las otras tres. Las celdas usan además esta matriz y este vector.



```python
B = Matrix([ [1,2,0], [0,1,3], [2,2,1] ])
c = Vector([2, 1, -1])
```

Primera: $\Vect{b}\Mat{I}=\Vect{b}$.



```python
b * I(3) == b
```

Segunda: la fila $i$-ésima de la identidad por una matriz es la fila $i$-ésima de la matriz.



```python
(2|I(3)) * A == 2|A
```

Tercera: $(\Vect{b}+\Vect{c})\Mat{A}=\Vect{b}\Mat{A}+\Vect{c}\Mat{A}$.



Cuarta: $(\lambda\Vect{b})\Mat{A}=\lambda(\Vect{b}\Mat{A})$.



Quinta: $\Vect{b}(\lambda\Mat{A})=(\lambda\Vect{b})\Mat{A}$.



```python
b * (3*A) == (3*b) * A
```

Sexta: $\Vect{b}(\Mat{A}+\Mat{B})=\Vect{b}\Mat{A}+\Vect{b}\Mat{B}$.



El libro deduce las seis de las propiedades del producto de una matriz por un vector. Usa la igualdad $\Vect{b}\Mat{A}=(\MatT{A})\Vect{b}$.



## Qué debe saber al terminar



-   El producto $\Vect{b}\Mat{A}$ es una combinación lineal de las filas de $\Mat{A}$. Los coeficientes son las componentes de $\Vect{b}$.
-   $\Mat{A}\Vect{b}$ y $\Vect{b}\Mat{A}$ son en general vectores distintos. Si la matriz no es cuadrada, con un mismo vector solo puede existir uno de los dos productos.
-   $\Vect{b}\Mat{A}=(\MatT{A})\Vect{b}$ y $\Mat{A}\Vect{b}=\Vect{b}(\MatT{A})$.
-   La componente $j$-ésima de $\Vect{b}\Mat{A}$ es el producto punto de $\Vect{b}$ por la columna $j$-ésima de $\Mat{A}$.
-   El producto de un vector por una matriz es lineal en el vector y lineal en la matriz.
